## 🔄 Part 2: While Loops - Continuous Processing

### Why Do We Need While Loops?

Imagine a sales system that processes data all day:
- A cashier rings up multiple customers (we don't know how many in advance)
- A warehouse receives deliveries throughout the day
- A customer browses products until they decide to checkout

**For loops** are great when you know the exact number of repetitions.
**While loops** are perfect when you need to repeat until a condition changes.

### While Loop Conceptual Model

Think of a while loop as a security guard:
```
While the door is unlocked:
    - Check each person
    - Process their request
    - Continue monitoring

Once door is locked:
    - Stop checking
    - Go home
```

### Syntax and Flow
```python
while condition_is_true:
    # Do something
    # Eventually change the condition to False
    # Otherwise loop runs forever!
```

**Visual Flow:**
```
Start
  ↓
Check condition → False → Exit loop
  ↓ True
Execute code block
  ↓
Change variable
  ↓
(Go back to check condition)
```

### Example 4: Basic While Loop

````python
# Count from 1 to 5
counter = 1

while counter <= 5:
    print(f"Count: {counter}")
    counter = counter + 1  # CRITICAL: Change the condition!

print("Loop finished!")
````

## 🛡️ Part 3: Input Validation - Protecting Your Program

### Why Validation Matters

**Real-world scenario:**
```python
# ❌ DANGEROUS: No validation
age = int(input("Enter your age: "))
# What if user types "twenty" or leaves it blank?
# Program crashes!
```

**Better approach:**

````python
# ✅ SAFE: With validation
age_input = input("Enter your age: ")

while not age_input.isdigit():
    print("❌ Invalid input! Please enter a number.")
    age_input = input("Enter your age: ")

age = int(age_input)
print(f"✅ Age recorded: {age}")
````

### Common Validation Patterns



#### Pattern 1: Numeric Validation

````python
def get_valid_number(prompt):
    """Keep asking until user enters a valid number"""
    while True:
        try:
            value = float(input(prompt))
            return value
        except ValueError:
            print("❌ Please enter a valid number!")

# Usage
price = get_valid_number("Enter price: $")
````

#### Pattern 2: Menu Selection

````python
def get_menu_choice(options):
    """Ensure user selects a valid menu option"""
    print("\nPlease select an option:")
    for i, option in enumerate(options, 1):
        print(f"{i}. {option}")

    while True:
        choice = input("Enter your choice (1-" + str(len(options)) + "): ")

        if choice.isdigit():
            choice_num = int(choice)
            if 1 <= choice_num <= len(options):
                return choice_num

        print(f"❌ Invalid choice! Please enter 1-{len(options)}")

# Usage
menu = ["View Sales", "Add Product", "Exit"]
selection = get_menu_choice(menu)
````

#### Pattern 3: Yes/No Confirmation

````python
def get_yes_no(prompt):
    """Get a yes or no response from user"""
    while True:
        response = input(prompt + " (yes/no): ").lower()

        if response in ['yes', 'y']:
            return True
        elif response in ['no', 'n']:
            return False
        else:
            print("❌ Please answer 'yes' or 'no'")

# Usage
if get_yes_no("Do you want to continue?"):
    print("Continuing...")
else:
    print("Exiting...")
````

## 💼 Part 4: Business Case Solution - Interactive Sales System

### Problem Statement

**MiniTech Electronics** needs a system where:

### Complete Implementation

````python
"""
MINITECH ELECTRONICS - INTERACTIVE SALES TRACKING SYSTEM
Module 1: Using input() and while loops
"""

print("=" * 60)
print("🏪 MINITECH ELECTRONICS SALES TRACKER")
print("=" * 60)
print("\nWelcome! This system tracks sales across all channels.")
print("You can enter as many sales transactions as needed.\n")

# ============================================
# DATA STRUCTURES (we'll learn more next week)
# ============================================

# Valid options for our business
VALID_CHANNELS = ["online", "physical store", "dealer"]
VALID_PRODUCTS = ["phone", "headphones", "charger", "tablet"]

# Storage for all sales transactions
all_sales = []
total_revenue = 0
transaction_count = 0

# ============================================
# HELPER FUNCTIONS FOR INPUT VALIDATION
# ============================================

def get_valid_choice(prompt, valid_options):
    """
    Keep asking until user provides a valid option

    Args:
        prompt: Question to ask user
        valid_options: List of acceptable answers

    Returns:
        Valid choice selected by user
    """
    print(f"\n{prompt}")
    for i, option in enumerate(valid_options, 1):
        print(f"  {i}. {option.title()}")

    while True:
        choice = input(f"Enter choice (1-{len(valid_options)}): ")

        if choice.isdigit():
            choice_num = int(choice)
            if 1 <= choice_num <= len(valid_options):
                return valid_options[choice_num - 1]

        print(f"❌ Invalid! Please enter a number between 1 and {len(valid_options)}")


def get_valid_number(prompt, allow_decimals=False):
    """
    Keep asking until user provides a valid number

    Args:
        prompt: Question to ask user
        allow_decimals: True if decimals are allowed, False for integers only

    Returns:
        Valid number entered by user
    """
    while True:
        value = input(f"\n{prompt}: ")

        try:
            if allow_decimals:
                number = float(value)
            else:
                if '.' in value:
                    print("❌ Please enter a whole number (no decimals)")
                    continue
                number = int(value)

            if number < 0:
                print("❌ Please enter a positive number")
                continue

            return number

        except ValueError:
            print("❌ Invalid input! Please enter a valid number")


def get_yes_no(prompt):
    """
    Get yes or no response from user

    Args:
        prompt: Question to ask

    Returns:
        True for yes, False for no
    """
    while True:
        response = input(f"\n{prompt} (yes/no): ").lower().strip()

        if response in ['yes', 'y']:
            return True
        elif response in ['no', 'n']:
            return False
        else:
            print("❌ Please answer 'yes' or 'no'")


# ============================================
# MAIN SALES ENTRY LOOP
# ============================================

continue_entry = True

while continue_entry:
    print("\n" + "=" * 60)
    print(f"TRANSACTION #{transaction_count + 1}")
    print("=" * 60)

    # Step 1: Get sales channel
    channel = get_valid_choice("Which sales channel?", VALID_CHANNELS)

    # Step 2: Get product
    product = get_valid_choice("Which product was sold?", VALID_PRODUCTS)

    # Step 3: Get quantity
    quantity = get_valid_number("Enter quantity sold", allow_decimals=False)

    # Step 4: Get unit price
    unit_price = get_valid_number("Enter unit price ($)", allow_decimals=True)

    # Step 5: Calculate transaction total
    transaction_total = quantity * unit_price

    # Step 6: Store the transaction
    transaction = {
        'channel': channel,
        'product': product,
        'quantity': quantity,
        'unit_price': unit_price,
        'total': transaction_total
    }
    all_sales.append(transaction)

    # Step 7: Update totals
    total_revenue += transaction_total
    transaction_count += 1

    # Step 8: Show confirmation
    print("\n" + "-" * 60)
    print("✅ TRANSACTION RECORDED")
    print("-" * 60)
    print(f"Channel: {channel.title()}")
    print(f"Product: {product.title()}")
    print(f"Quantity: {quantity} units")
    print(f"Unit Price: ${unit_price:.2f}")
    print(f"Transaction Total: ${transaction_total:.2f}")
    print("-" * 60)

    # Step 9: Ask if user wants to enter more
    continue_entry = get_yes_no("Would you like to enter another transaction?")


# ============================================
# GENERATE SUMMARY REPORT
# ============================================

print("\n\n" + "=" * 60)
print("📊 DAILY SALES SUMMARY REPORT")
print("=" * 60)

if transaction_count == 0:
    print("\n⚠️  No transactions were recorded today.")
else:
    print(f"\nTotal Transactions: {transaction_count}")
    print(f"Total Revenue: ${total_revenue:.2f}")
    print(f"Average Transaction: ${(total_revenue/transaction_count):.2f}")

    # Break down by channel
    print("\n" + "-" * 60)
    print("BREAKDOWN BY SALES CHANNEL:")
    print("-" * 60)

    for channel in VALID_CHANNELS:
        channel_sales = [sale for sale in all_sales if sale['channel'] == channel]
        channel_count = len(channel_sales)
        channel_revenue = sum(sale['total'] for sale in channel_sales)

        if channel_count > 0:
            print(f"\n{channel.title()}:")
            print(f"  Transactions: {channel_count}")
            print(f"  Revenue: ${channel_revenue:.2f}")

    # Break down by product
    print("\n" + "-" * 60)
    print("BREAKDOWN BY PRODUCT:")
    print("-" * 60)

    for product in VALID_PRODUCTS:
        product_sales = [sale for sale in all_sales if sale['product'] == product]
        product_quantity = sum(sale['quantity'] for sale in product_sales)
        product_revenue = sum(sale['total'] for sale in product_sales)

        if product_quantity > 0:
            print(f"\n{product.title()}:")
            print(f"  Units Sold: {product_quantity}")
            print(f"  Revenue: ${product_revenue:.2f}")

print("\n" + "=" * 60)
print("Thank you for using MiniTech Sales Tracker!")
print("=" * 60)
````

## ❓ Common Questions & Mistakes

### Q1: "My while loop runs forever! How do I stop it?"
**A:** This is called an "infinite loop". Make sure:
```python
# ❌ WRONG: Condition never changes
count = 1
while count < 5:
    print(count)
    # Forgot to increment count!

# ✅ CORRECT: Condition eventually becomes False
count = 1
while count < 5:
    print(count)
    count += 1  # This will eventually make condition False
```

### Q2: "Why do I get ValueError when converting input?"
**A:** User typed something that can't be converted:
```python
# ❌ DANGEROUS
age = int(input("Age: "))  # Crashes if user types "twenty"

# ✅ SAFE
while True:
    try:
        age = int(input("Age: "))
        break  # Exit loop if successful
    except ValueError:
        print("Please enter a number!")
```

### Q3: "How do I know when to use for vs while?"
**A:**
- **Use for** when you know how many times to repeat
  - "Process all 10 products"
  - "Check each item in this list"
- **Use while** when repetitions depend on a condition
  - "Keep asking until user enters 'quit'"
  - "Continue until total reaches $1000"

### Creating Dictionaries

**Method 1: Direct Creation**

````python
 Empty dictionary
product = {}

# Dictionary with initial data
product = {
    "name": "iPhone 15",
    "price": 699.99,
    "stock": 45,
    "category": "Phone"
}
````

**Method 2: Building Gradually**

````python
product = {}
product["name"] = "iPhone 15"
product["price"] = 699.99
product["stock"] = 45
product["category"] = "Phone"
````

### Accessing Dictionary Data

````python
product = {
    "name": "iPhone 15",
    "price": 699.99,
    "stock": 45
}

# Access by key
print(product["name"])     # Output: iPhone 15
print(product["price"])    # Output: 699.99

# Modify values
product["stock"] = 40      # Reduce stock
product["price"] = 649.99  # Update price

# Add new keys
product["supplier"] = "Apple Inc."

print(product)
# Output: {'name': 'iPhone 15', 'price': 649.99, 'stock': 40, 'supplier': 'Apple Inc.'}
````

### Safe Access with `.get()`


```python
product = {"name": "iPhone", "price": 699.99}

# ❌ DANGEROUS: KeyError if key doesn't exist
print(product["stock"])  # ERROR! Key 'stock' doesn't exist


```

````python
# ✅ SAFE: Returns None if key doesn't exist
print(product.get("stock"))  # Output: None

# ✅ EVEN BETTER: Provide default value
print(product.get("stock", 0))  # Output: 0
````

### Common Dictionary Operations

````python
product = {
    "name": "iPhone 15",
    "price": 699.99,
    "stock": 45,
    "category": "Phone"
}

# Check if key exists
if "price" in product:
    print("Price is available")

# Get all keys
keys = product.keys()
print(keys)  # dict_keys(['name', 'price', 'stock', 'category'])

# Get all values
values = product.values()
print(values)  # dict_values(['iPhone 15', 699.99, 45, 'Phone'])

# Get all key-value pairs
items = product.items()
print(items)
# dict_items([('name', 'iPhone 15'), ('price', 699.99), ...])

# Delete a key
del product["category"]

# Remove and return value
removed_price = product.pop("price")
print(removed_price)  # 699.99
````

### Iterating Through Dictionaries

````python
product = {"name": "iPhone", "price": 699.99, "stock": 45}

# Method 1: Iterate through keys
for key in product:
    print(f"{key}: {product[key]}")

# Method 2: Iterate through key-value pairs (better!)
for key, value in product.items():
    print(f"{key}: {value}")
````

## 📦 Part 3: Tuples - Immutable Sequences

### What Are Tuples?

**Conceptual Understanding:**
- Tuples are like lists, but **immutable** (cannot be changed)
- Once created, you cannot add, remove, or modify elements
- Perfect for data that should never change

**Why Use Tuples?**
1. **Data Protection**: Prevent accidental modifications
2. **Performance**: Slightly faster than lists
3. **Dictionary Keys**: Can be used as dictionary keys (lists cannot)
4. **Clear Intent**: Signals to other programmers: "Don't change this!"

### Creating Tuples


````python
# Empty tuple
empty = ()

# Tuple with elements
coordinates = (40.7128, -74.0060)  # New York City coordinates
rgb_color = (255, 0, 128)          # Color values

# Single element tuple (note the comma!)
single = (42,)    # ✅ Correct
not_tuple = (42)  # ❌ This is just an integer!

# Creating from list
my_list = [1, 2, 3]
my_tuple = tuple(my_list)
print(my_tuple)
````

### Accessing Tuple Data

````python
coordinates = (40.7128, -74.0060)

# Access by index (just like lists)
latitude = coordinates[0]   # 40.7128
longitude = coordinates[1]  # -74.0060

# Unpacking tuples (very useful!)
lat, lon = coordinates
print(f"Latitude: {lat}, Longitude: {lon}")

# Slicing
numbers = (10, 20, 30, 40, 50)
subset = numbers[1:4]  # (20, 30, 40)
````

### Tuples Are Immutable

```python
product_info = ("iPhone", 699.99, "Phone")

# ❌ THESE ALL CAUSE ERRORS:
# product_info[1] = 649.99      # Can't modify
# product_info.append("Apple")  # Can't add
# del product_info[0]           # Can't delete

# ✅ MUST CREATE NEW TUPLE:
product_info = ("iPhone", 649.99, "Phone")  # Completely new tuple
```


| Use TUPLE when: | Use LIST when: |
|-----------------|----------------|
| Data shouldn't change | Data will be modified |
| Fixed structure (e.g., coordinates) | Size may change |
| Dictionary keys needed | Not used as keys |
| Function returns multiple values | Single collection of items |

**Examples:**
```python
# ✅ TUPLE: Geographic coordinates don't change

# ✅ LIST: Shopping cart items change
shopping_cart = ["Phone", "Charger"]

# ✅ TUPLE: RGB values are fixed

# ✅ LIST: Inventory changes
```

````python
"""
MINITECH ELECTRONICS - ADVANCED SALES TRACKING SYSTEM
Module 2: Using Dictionaries and Tuples
"""

import datetime

print("=" * 70)
print("🏪 MINITECH ELECTRONICS - ADVANCED SALES TRACKER")
print("=" * 70)

# ============================================
# REFERENCE DATA (Tuples - Immutable)
# ============================================

# Product codes that never change
PRODUCT_CODES = (
    ("PHN", "Phone"),
    ("HDP", "Headphones"),
    ("CHG", "Charger"),
    ("TAB", "Tablet")
)

# Sales channels (order matters for reporting)
SALES_CHANNELS = ("Online Store", "Physical Store", "Authorized Dealer")

# Tax rate (fixed by law)
TAX_RATE = 0.08  # 8%

# ============================================
# PRODUCT CATALOG (Dictionary)
# ============================================

product_catalog = {
    "PHN001": {
        "name": "iPhone 15 Pro",
        "base_price": 999.99,
        "stock": 45,
        "category": "Phone",
        "specs": {
            "storage": "256GB",
            "color": "Titanium Blue"
        }
    },
    "PHN002": {
        "name": "Samsung Galaxy S24",
        "base_price": 899.99,
        "stock": 38,
        "category": "Phone",
        "specs": {
            "storage": "256GB",
            "color": "Phantom Black"
        }
    },
    "HDP001": {
        "name": "AirPods Pro 2",
        "base_price": 249.99,
        "stock": 120,
        "category": "Headphones",
        "specs": {
            "type": "In-ear",
            "noise_cancellation": True
        }
    },
    "HDP002": {
        "name": "Sony WH-1000XM5",
        "base_price": 399.99,
        "stock": 75,
        "category": "Headphones",
        "specs": {
            "type": "Over-ear",
            "noise_cancellation": True
        }
    },
    "CHG001": {
        "name": "USB-C Fast Charger",
        "base_price": 29.99,
        "stock": 200,
        "category": "Charger",
        "specs": {
            "watts": "20W",
            "cable_included": True
        }
    },
    "TAB001": {
        "name": "iPad Air",
        "base_price": 599.99,
        "stock": 55,
        "category": "Tablet",
        "specs": {
            "screen": "10.9 inch",
            "storage": "64GB"
        }
    }
}

# ============================================
# SALES TRANSACTIONS STORAGE
# ============================================

# List of dictionaries - each sale is a complete record
sales_transactions = []
transaction_counter = 1

# ============================================
# HELPER FUNCTIONS
# ============================================

def display_product_catalog():
    """Show all available products"""
    print("\n" + "=" * 70)
    print("📦 AVAILABLE PRODUCTS")
    print("=" * 70)

    for product_id, details in product_catalog.items():
        print(f"\nCode: {product_id}")
        print(f"  Name: {details['name']}")
        print(f"  Price: ${details['base_price']:.2f}")
        print(f"  Stock: {details['stock']} units")
        print(f"  Category: {details['category']}")


def get_product_choice():
    """Let user select a product with validation"""
    while True:
        product_id = input("\nEnter product code (e.g., PHN001): ").upper()

        if product_id in product_catalog:
            product = product_catalog[product_id]
            if product['stock'] > 0:
                return product_id
            else:
                print(f"❌ {product['name']} is out of stock!")
        else:
            print("❌ Invalid product code!")


def get_channel_choice():
    """Let user select sales channel"""
    print("\n📍 SELECT SALES CHANNEL:")
    for i, channel in enumerate(SALES_CHANNELS, 1):
        print(f"  {i}. {channel}")

    while True:
        choice = input(f"Enter choice (1-{len(SALES_CHANNELS)}): ")
        if choice.isdigit():
            choice_num = int(choice)
            if 1 <= choice_num <= len(SALES_CHANNELS):
                return SALES_CHANNELS[choice_num - 1]
        print(f"❌ Please enter 1-{len(SALES_CHANNELS)}")


def get_quantity(product_id):
    """Get valid quantity that doesn't exceed stock"""
    max_stock = product_catalog[product_id]['stock']

    while True:
        try:
            quantity = int(input(f"\nEnter quantity (max {max_stock}): "))
            if 1 <= quantity <= max_stock:
                return quantity
            else:
                print(f"❌ Please enter 1-{max_stock}")
        except ValueError:
            print("❌ Please enter a valid number!")


def process_sale(product_id, quantity, channel):
    """
    Process a sale transaction and return transaction record

    Returns: Dictionary with complete transaction details
    """
    global transaction_counter

    product = product_catalog[product_id]

    # Calculate financial details
    unit_price = product['base_price']
    subtotal = unit_price * quantity
    tax_amount = subtotal * TAX_RATE
    total = subtotal + tax_amount

    # Create transaction record
    transaction = {
        "transaction_id": f"TXN{transaction_counter:04d}",
        "timestamp": datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "product_id": product_id,
        "product_name": product['name'],
        "channel": channel,
        "quantity": quantity,
        "unit_price": unit_price,
        "subtotal": subtotal,
        "tax": tax_amount,
        "total": total
    }

    # Update stock
    product_catalog[product_id]['stock'] -= quantity

    # Increment counter
    transaction_counter += 1

    return transaction


def display_transaction(transaction):
    """Show transaction receipt"""
    print("\n" + "=" * 70)
    print("🧾 TRANSACTION RECEIPT")
    print("=" * 70)
    print(f"Transaction ID: {transaction['transaction_id']}")
    print(f"Date/Time: {transaction['timestamp']}")
    print(f"Channel: {transaction['channel']}")
    print(f"\nProduct: {transaction['product_name']} ({transaction['product_id']})")
    print(f"Quantity: {transaction['quantity']} units")
    print(f"Unit Price: ${transaction['unit_price']:.2f}")
    print(f"\nSubtotal: ${transaction['subtotal']:.2f}")
    print(f"Tax ({TAX_RATE*100:.0f}%): ${transaction['tax']:.2f}")
    print(f"{'─'*70}")
    print(f"TOTAL: ${transaction['total']:.2f}")
    print("=" * 70)


def generate_summary_report():
    """Generate comprehensive sales report"""
    print("\n\n" + "=" * 70)
    print("📊 SALES SUMMARY REPORT")
    print("=" * 70)

    if not sales_transactions:
        print("\n⚠️  No transactions recorded.")
        return

    # Overall statistics
    total_transactions = len(sales_transactions)
    total_revenue = sum(t['total'] for t in sales_transactions)
    total_tax_collected = sum(t['tax'] for t in sales_transactions)
    avg_transaction = total_revenue / total_transactions

    print(f"\n📈 OVERALL STATISTICS:")
    print(f"  Total Transactions: {total_transactions}")
    print(f"  Total Revenue: ${total_revenue:.2f}")
    print(f"  Tax Collected: ${total_tax_collected:.2f}")
    print(f"  Average Transaction: ${avg_transaction:.2f}")

    # Breakdown by channel
    print(f"\n🏬 BREAKDOWN BY CHANNEL:")
    channel_stats = {}

    for channel in SALES_CHANNELS:
        channel_sales = [t for t in sales_transactions if t['channel'] == channel]
        if channel_sales:
            count = len(channel_sales)
            revenue = sum(t['total'] for t in channel_sales)
            channel_stats[channel] = (count, revenue)

            print(f"\n  {channel}:")
            print(f"    Transactions: {count}")
            print(f"    Revenue: ${revenue:.2f}")
            print(f"    Avg Transaction: ${revenue/count:.2f}")

    # Breakdown by product category
    print(f"\n📦 BREAKDOWN BY CATEGORY:")
    category_stats = {}

    for transaction in sales_transactions:
        product_id = transaction['product_id']
        category = product_catalog[product_id]['category']

        if category not in category_stats:
            category_stats[category] = {
                'units': 0,
                'revenue': 0
            }

        category_stats[category]['units'] += transaction['quantity']
        category_stats[category]['revenue'] += transaction['total']

    for category, stats in sorted(category_stats.items()):
        print(f"\n  {category}:")
        print(f"    Units Sold: {stats['units']}")
        print(f"    Revenue: ${stats['revenue']:.2f}")

    # Top selling products
    print(f"\n🏆 TOP SELLING PRODUCTS:")
    product_sales = {}

    for transaction in sales_transactions:
        prod_id = transaction['product_id']
        if prod_id not in product_sales:
            product_sales[prod_id] = {
                'name': transaction['product_name'],
                'units': 0,
                'revenue': 0
            }

        product_sales[prod_id]['units'] += transaction['quantity']
        product_sales[prod_id]['revenue'] += transaction['total']

    # Sort by revenue
    sorted_products = sorted(
        product_sales.items(),
        key=lambda x: x[1]['revenue'],
        reverse=True
    )

    for i, (prod_id, stats) in enumerate(sorted_products[:3], 1):
        print(f"\n  #{i} {stats['name']} ({prod_id}):")
        print(f"     Units: {stats['units']}")
        print(f"     Revenue: ${stats['revenue']:.2f}")


# ============================================
# MAIN PROGRAM
# ============================================

def main():
    """Main program loop"""
    print("\nWelcome to the Advanced Sales Tracking System!")
    print("This system helps you record and analyze sales transactions.\n")

    while True:
        print("\n" + "=" * 70)
        print("MAIN MENU")
        print("=" * 70)
        print("1. View Product Catalog")
        print("2. Record New Sale")
        print("3. View All Transactions")
        print("4. Generate Summary Report")
        print("5. Exit")

        choice = input("\nEnter your choice (1-5): ")

        if choice == "1":
            display_product_catalog()

        elif choice == "2":
            display_product_catalog()
            product_id = get_product_choice()
            quantity = get_quantity(product_id)
            channel = get_channel_choice()

            transaction = process_sale(product_id, quantity, channel)
            sales_transactions.append(transaction)
            display_transaction(transaction)

        elif choice == "3":
            if not sales_transactions:
                print("\n⚠️  No transactions to display.")
            else:
                print("\n" + "=" * 70)
                print("ALL TRANSACTIONS")
                print("=" * 70)
                for t in sales_transactions:
                    print(f"\n{t['transaction_id']} | {t['timestamp']}")
                    print(f"  {t['product_name']} x{t['quantity']}")
                    print(f"  Channel: {t['channel']}")
                    print(f"  Total: ${t['total']:.2f}")

        elif choice == "4":
            generate_summary_report()

        elif choice == "5":
            print("\n" + "=" * 70)
            print("Thank you for using MiniTech Sales Tracker!")
            print("=" * 70)
            break

        else:
            print("❌ Invalid choice! Please enter 1-5")


# Run the program
if __name__ == "__main__":
    main()
````

## 🎯 Understanding the Solution

### Key Design Decisions

1. **Tuples for Constants:**
```python
PRODUCT_CODES = (("PHN", "Phone"), ...)  # Won't change
SALES_CHANNELS = ("Online", "Physical", "Dealer")  # Fixed list
TAX_RATE = 0.08  # Set by law
```

2. **Dictionaries for Dynamic Data:**
```python
product_catalog = {
    "PHN001": {
        "name": "iPhone",
        "stock": 45  # Changes with sales
    }
}
```

3. **List of Dictionaries for Records:**
```python
sales_transactions = [
    {"transaction_id": "TXN001", ...},
    {"transaction_id": "TXN002", ...}
]
```


## ❓ Common Questions

### Q1: "When should I use a dictionary vs a list?"
**A:**
- **Dictionary**: When you need named access ("price", "name")
- **List**: When order matters and items are similar ([1,2,3])

### Q2: "Can I modify a tuple?"
**A:** No! That's the point. If you need to change data, use a list.

### Q3: "What's the difference between dict.get() and dict[]?"
**A:**
- `dict["key"]` - Raises error if key doesn't exist
- `dict.get("key")` - Returns None (or default) if key doesn't exist

## 📝 Key Takeaways

✅ **Dictionaries:**
- Store data with meaningful keys
- Perfect for structured records
- Use `.get()` for safe access
- Can be nested for complex data

✅ **Tuples:**
- Immutable sequences
- Protect data from changes
- Slightly faster than lists
- Good for fixed data

✅ **Best Practices:**
- Use tuples for constants
- Use dictionaries for structured data
- Combine them for powerful solutions
- Choose based on mutability needs
