## 1. Introduction to File Operations

### Why File Operations Matter in Business

In real-world applications, data doesn't just exist in memory during program execution. Businesses need to:

- **Persist data**: Save information permanently (sales records, inventory, customer data)
- **Share data**: Exchange information between systems and departments
- **Analyze data**: Process historical data for insights and reporting
- **Backup data**: Maintain data integrity and disaster recovery
- **Audit trails**: Keep logs of transactions and operations

**Real-world scenarios:**
- A retail system saves daily sales to a file at closing time
- An inventory system reads product data from a CSV file
- A financial system writes transaction logs for auditing
- A reporting system generates weekly sales summaries

### File Operations Overview

**Three fundamental operations:**

1. **Opening a file**: Establish connection between program and file
2. **Reading/Writing**: Transfer data between program and file
3. **Closing a file**: Release system resources and ensure data is saved

**Key Concepts:**

- **File handle/object**: Python object that represents the open file
- **File path**: Location of file in the file system (absolute or relative)
- **File modes**: How we want to interact with the file (read, write, append)
- **Buffer**: Temporary memory storage before writing to disk
- **Text vs Binary**: Files contain either text (human-readable) or binary data

## 2. File Opening Modes

### Understanding File Modes

When opening a file, you must specify how you intend to use it. Python provides several modes:

| Mode | Description | Creates File | Overwrites | Pointer Position |
|------|-------------|--------------|------------|------------------|
| `'r'` | Read (default) | No | No | Beginning |
| `'w'` | Write | Yes | Yes (destroys existing content) | Beginning |
| `'a'` | Append | Yes | No (adds to end) | End |
| `'r+'` | Read and Write | No | No | Beginning |
| `'w+'` | Write and Read | Yes | Yes | Beginning |
| `'a+'` | Append and Read | Yes | No | End |
| `'rb'` | Read Binary | No | No | Beginning |
| `'wb'` | Write Binary | Yes | Yes | Beginning |

**Best Practices:**

- **Use `'r'` mode** when you only need to read data (safest, prevents accidental changes)
- **Use `'w'` mode** carefully - it erases existing content!
- **Use `'a'` mode** for log files and continuous data collection
- **Use text mode** (default) for .txt, .csv, .json files
- **Use binary mode** (`'b'`) for images, audio, video, or binary data files

**Mode Selection Decision Tree:**

```
Need to read file?
├─ Yes → Need to modify?
│  ├─ Yes → Use 'r+'
│  └─ No → Use 'r' (safest)
└─ No → Need to preserve existing content?
   ├─ Yes → Use 'a' (append)
   └─ No → Use 'w' (overwrites)
```

# 3. Basic File Operations

### Opening and Closing Files (Traditional Method)

The traditional approach requires manual file closing:
```python
# Opening a file
file_handle = open('filename.txt', 'r')

# Do something with the file
content = file_handle.read()

# MUST close the file
file_handle.close()
```

**Problems with this approach:**

1. **Easy to forget** to close the file
2. **Resource leaks** if exception occurs before closing
3. **Data corruption** if program crashes before close()
4. **Manual error handling** required

### The Modern Way: Context Managers (with statement)

**Python's recommended approach** uses the `with` statement:

```python
with open('filename.txt', 'r') as file_handle:
    content = file_handle.read()
    # File automatically closes when block ends
# File is guaranteed to be closed here, even if error occurred
```

**Benefits:**

- ✅ Automatically closes file when done
- ✅ Closes file even if exception occurs
- ✅ Cleaner, more readable code
- ✅ Follows Python best practices (PEP 343)
- ✅ Prevents resource leaks

**Always use `with` statement for file operations!**


## 4. Reading Files

### Reading Methods

Python provides several methods to read file content:

| Method | Description | Returns | Use Case |
|--------|-------------|---------|----------|
| `.read()` | Read entire file | String | Small files that fit in memory |
| `.read(size)` | Read specific number of characters | String | Large files, read in chunks |
| `.readline()` | Read one line | String | Processing line by line |
| `.readlines()` | Read all lines | List of strings | Small files, need all lines |
| Iteration | `for line in file` | Iterates strings | Large files, memory efficient |


**Method Comparison:**

```python
# Method 1: read() - entire file as single string
with open('data.txt', 'r') as file:
    content = file.read()
    # content = "line1\nline2\nline3\n"

# Method 2: readline() - one line at a time
with open('data.txt', 'r') as file:
    line1 = file.readline()  # "line1\n"
    line2 = file.readline()  # "line2\n"

# Method 3: readlines() - list of all lines
with open('data.txt', 'r') as file:
    lines = file.readlines()
    # lines = ["line1\n", "line2\n", "line3\n"]

# Method 4: Iteration - most Pythonic for large files
with open('data.txt', 'r') as file:
    for line in file:
        process(line)  # Memory efficient
```

**When to use each method:**

- **`.read()`**: Configuration files, small data files, when you need entire content
- **`.readline()`**: When you need to process lines individually with complex logic
- **`.readlines()`**: When you need all lines as a list and file is small
- **Iteration**: Large files, log files, streaming data (RECOMMENDED for most cases)

## 5. Writing Files

### Writing Methods

| Method | Description | Adds Newline | Returns |
|--------|-------------|--------------|---------|
| `.write(string)` | Write string to file | No | Number of characters written |
| `.writelines(list)` | Write list of strings | No | None |
| `print(..., file=f)` | Write with print function | Yes (default) | None |

**Important Notes:**

1. **`.write()` doesn't add newlines** - you must add `\n` manually
2. **`.writelines()` doesn't add newlines** - list items must include `\n`
3. **`print()` adds newline by default** - set `end=''` to disable
4. **Write returns count** - useful for verification
5. **Must open in write mode** - `'w'`, `'a'`, or `'r+'`

### Write vs Append Mode

**Write mode (`'w'`):** Destroys existing content

```python
# First write
with open('output.txt', 'w') as file:
    file.write('First line\n')
# File contains: "First line\n"

# Second write - OVERWRITES!
with open('output.txt', 'w') as file:
    file.write('Second line\n')
# File contains: "Second line\n" (first line is gone!)
```

**Append mode (`'a'`):** Adds to existing content

```python
# First write
with open('log.txt', 'a') as file:
    file.write('First entry\n')
# File contains: "First entry\n"

# Second write - APPENDS
with open('log.txt', 'a') as file:
    file.write('Second entry\n')
# File contains: "First entry\nSecond entry\n"
```

**Common Pattern: Creating Files if Doesn't Exist**

```python
# Append mode creates file if it doesn't exist
with open('new_file.txt', 'a') as file:
    file.write('This creates the file if needed\n')

## 6. Working with CSV Files

### Why CSV Files Matter in Business

**CSV (Comma-Separated Values)** is the most common format for business data exchange:

- ✅ **Universal compatibility**: Excel, databases, analytics tools all support CSV
- ✅ **Human-readable**: Can open in any text editor
- ✅ **Lightweight**: No complex formatting, just data
- ✅ **Easy to generate**: Simple structure, easy to program
- ✅ **Industry standard**: Sales data, inventory, customer lists

**CSV Structure:**

```
ProductID,ProductName,Price,Stock
SKU_01,iPhone 13,699.99,45
SKU_02,Samsung Galaxy S22,799.99,32
SKU_03,MacBook Pro,1299.99,18
```

- **First line**: Header (column names)
- **Subsequent lines**: Data rows
- **Delimiter**: Comma (sometimes semicolon or tab)
- **Text qualifiers**: Quotes for text containing commas

### Python's csv Module

Python provides a built-in `csv` module for robust CSV handling:

**Advantages over manual parsing:**

- ✅ Handles edge cases (commas in data, quotes, newlines)
- ✅ Automatic type handling
- ✅ Dictionary-based access
- ✅ Robust and tested
- ✅ Memory efficient for large files

**Basic CSV Operations:**

````python
import csv

# Reading CSV
with open('data.csv', 'r') as file:
    csv_reader = csv.reader(file)
    for row in csv_reader:
        print(row)  # row is a list

# Writing CSV
with open('output.csv', 'w', newline='') as file:
    csv_writer = csv.writer(file)
    csv_writer.writerow(['Name', 'Age', 'City'])
    csv_writer.writerow(['Alice', 25, 'New York'])
````

**Important: `newline=''` parameter**

Always use `newline=''` when opening CSV files in write mode to prevent extra blank lines on Windows.


## 7. Error Handling in File Operations

### Common File Errors

| Exception | Cause | Solution |
|-----------|-------|----------|
| `FileNotFoundError` | File doesn't exist (read mode) | Check file exists, use correct path |
| `PermissionError` | No permission to access file | Check file permissions |
| `IsADirectoryError` | Tried to open directory as file | Verify path points to file |
| `OSError` | System-related error | Check disk space, file locks |
| `UnicodeDecodeError` | Wrong encoding | Specify encoding parameter |

### Proper Error Handling

**Basic pattern:**

```python
try:
    with open('data.txt', 'r') as file:
        content = file.read()
except FileNotFoundError:
    print("Error: File not found")
except PermissionError:
    print("Error: No permission to read file")
except Exception as e:
    print(f"Unexpected error: {e}")
```

**Checking file existence:**

```python
import os

if os.path.exists('data.txt'):
    with open('data.txt', 'r') as file:
        content = file.read()
else:
    print("File does not exist")
```

**Best practices:**

- Always use `with` statement (handles closing even with errors)
- Catch specific exceptions first, general Exception last
- Provide meaningful error messages to users
- Log errors for debugging
- Have fallback behavior when file operations fail

**Relative path:** Path relative to current working directory
```

## 9. Best Practices for File Operations

### General Guidelines

1. **Always use `with` statement**
   - Guarantees file closure
   - Handles exceptions gracefully
   - More readable code

2. **Choose appropriate file mode**
   - Use `'r'` when only reading (safest)
   - Be careful with `'w'` (overwrites)
   - Use `'a'` for logs and incremental data

3. **Handle errors appropriately**
   - Use try-except blocks
   - Provide meaningful error messages
   - Don't let file errors crash your program

4. **Use CSV module for structured data**
   - Don't parse CSV manually
   - Handles edge cases correctly
   - More robust and maintainable

5. **Close files in the correct order**
   - If you open multiple files, close in reverse order
   - `with` statement handles this automatically

6. **Consider encoding**
   - Specify encoding explicitly: `encoding='utf-8'`
   - Important for international characters
   - Prevents UnicodeDecodeError

7. **Validate file operations**
   - Check if file exists before reading
   - Verify write success
   - Use os.path or pathlib for path operations

### Performance Considerations

**For large files:**

```python
# ❌ BAD: Reads entire file into memory
with open('huge_file.txt', 'r') as file:
    content = file.read()  # May crash if file is huge

# ✅ GOOD: Process line by line
with open('huge_file.txt', 'r') as file:
    for line in file:
        process(line)  # Memory efficient
```

**For frequent writes:**

```python
# ❌ BAD: Opens/closes file repeatedly
for i in range(1000):
    with open('output.txt', 'a') as file:
        file.write(f'Line {i}\n')

# ✅ GOOD: Open once, write multiple times
with open('output.txt', 'a') as file:
    for i in range(1000):
        file.write(f'Line {i}\n')
```

## Code Implementation

### Step 1: Reading Product Catalog from CSV

````python
import csv
import os
from datetime import datetime

# ============================================
# STEP 1: LOAD PRODUCT CATALOG
# ============================================

def load_product_catalog(filename='products.csv'):
    """
    Load product catalog from CSV file into a dictionary.

    Args:
        filename (str): Path to products CSV file

    Returns:
        dict: Product catalog with ProductID as key
    """
    products = {}

    try:
        with open(filename, 'r', encoding='utf-8') as file:
            csv_reader = csv.DictReader(file)

            for row in csv_reader:
                product_id = row['ProductID']
                products[product_id] = {
                    'name': row['ProductName'],
                    'category': row['Category'],
                    'price': float(row['Price']),
                    'stock': int(row['Stock'])
                }

        print(f"✅ Loaded {len(products)} products from {filename}")
        return products

    except FileNotFoundError:
        print(f"❌ Error: {filename} not found")
        print("Creating empty product catalog...")
        return {}

    except KeyError as e:
        print(f"❌ Error: Missing column {e} in CSV file")
        return {}

    except ValueError as e:
        print(f"❌ Error: Invalid data format - {e}")
        return {}


# Example usage
print("=" * 60)
print("TECHSPHERE ELECTRONICS - PRODUCT CATALOG LOADER")
print("=" * 60)

catalog = load_product_catalog('products.csv')

if catalog:
    print("\n📦 Product Catalog:")
    for product_id, details in catalog.items():
        print(f"\n{product_id}: {details['name']}")
        print(f"  Category: {details['category']}")
        print(f"  Price: ${details['price']:.2f}")
        print(f"  Stock: {details['stock']} units")
````

### Step 3: Recording Sales Transactions

````python
# ============================================
# STEP 3: SALES TRANSACTION LOGGING
# ============================================

def record_sale(product_id, quantity, channel, catalog, log_file='sales_log.txt'):
    """
    Record a sale transaction to the log file.

    Args:
        product_id (str): Product identifier
        quantity (int): Number of units sold
        channel (str): Sales channel (Online, Store, Phone)
        catalog (dict): Product catalog
        log_file (str): Path to log file

    Returns:
        tuple: (success, total_amount)
    """
    # Validate product exists
    if product_id not in catalog:
        print(f"❌ Error: Product {product_id} not found in catalog")
        return False, 0.0

    product = catalog[product_id]

    # Check stock availability
    if product['stock'] < quantity:
        print(f"❌ Error: Insufficient stock for {product['name']}")
        print(f"   Requested: {quantity}, Available: {product['stock']}")
        return False, 0.0

    # Calculate total
    unit_price = product['price']
    total_amount = unit_price * quantity

    # Create log entry
    timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    log_entry = (f"{timestamp} | Sale | {product_id} | "
                 f"Qty: {quantity} | Total: ${total_amount:.2f} | {channel}\n")

    try:
        # Append to log file (creates if doesn't exist)
        with open(log_file, 'a', encoding='utf-8') as file:
            file.write(log_entry)

        # Update stock in catalog
        product['stock'] -= quantity

        print(f"✅ Sale recorded: {product['name']} x{quantity} = ${total_amount:.2f}")
        return True, total_amount

    except IOError as e:
        print(f"❌ Error writing to log file: {e}")
        return False, 0.0


# Example usage
print("\n" + "=" * 60)
print("RECORDING SALES TRANSACTIONS")
print("=" * 60)

# Simulate several sales
sales_to_process = [
    ('SKU_01', 2, 'Online'),
    ('SKU_03', 1, 'Store'),
    ('SKU_05', 3, 'Online'),
    ('SKU_02', 1, 'Phone'),
    ('SKU_08', 2, 'Store'),
]

total_revenue = 0.0
successful_sales = 0

for product_id, qty, channel in sales_to_process:
    success, amount = record_sale(product_id, qty, channel, catalog)
    if success:
        successful_sales += 1
        total_revenue += amount
    print()

print("=" * 60)
print(f"📊 Session Summary:")
print(f"   Successful sales: {successful_sales}")
print(f"   Total revenue: ${total_revenue:.2f}")
print("=" * 60)
````

### Step 4: Reading and Analyzing Sales Log


````python
# ============================================
# STEP 4: READING AND ANALYZING SALES LOG
# ============================================

def analyze_sales_log(log_file='sales_log.txt'):
    """
    Read and analyze sales log file.

    Args:
        log_file (str): Path to sales log file

    Returns:
        dict: Sales analytics
    """
    if not os.path.exists(log_file):
        print(f"❌ Log file {log_file} not found")
        return None

    sales_data = {
        'total_transactions': 0,
        'total_revenue': 0.0,
        'by_channel': {},
        'by_product': {}
    }

    try:
        with open(log_file, 'r', encoding='utf-8') as file:
            for line in file:
                # Parse log entry
                # Format: YYYY-MM-DD HH:MM:SS | Sale | SKU | Qty: N | Total: $X.XX | Channel
                parts = line.strip().split(' | ')

                if len(parts) >= 5:
                    timestamp = parts[0]
                    product_id = parts[2]
                    quantity = int(parts[3].split(': ')[1])
                    total_str = parts[4].split(': ')[1].replace('$', '')
                    total = float(total_str)
                    channel = parts[5]

                    # Update statistics
                    sales_data['total_transactions'] += 1
                    sales_data['total_revenue'] += total

                    # By channel
                    if channel not in sales_data['by_channel']:
                        sales_data['by_channel'][channel] = {
                            'count': 0,
                            'revenue': 0.0
                        }
                    sales_data['by_channel'][channel]['count'] += 1
                    sales_data['by_channel'][channel]['revenue'] += total

                    # By product
                    if product_id not in sales_data['by_product']:
                        sales_data['by_product'][product_id] = {
                            'units_sold': 0,
                            'revenue': 0.0
                        }
                    sales_data['by_product'][product_id]['units_sold'] += quantity
                    sales_data['by_product'][product_id]['revenue'] += total

        return sales_data

    except Exception as e:
        print(f"❌ Error analyzing log file: {e}")
        return None


# Analyze the sales log
print("\n" + "=" * 60)
print("SALES LOG ANALYSIS")
print("=" * 60)

analytics = analyze_sales_log()

if analytics:
    print(f"\n📊 Overall Statistics:")
    print(f"   Total Transactions: {analytics['total_transactions']}")
    print(f"   Total Revenue: ${analytics['total_revenue']:.2f}")

    if analytics['total_transactions'] > 0:
        avg_transaction = analytics['total_revenue'] / analytics['total_transactions']
        print(f"   Average Transaction: ${avg_transaction:.2f}")

    print(f"\n🏪 Sales by Channel:")
    for channel, stats in analytics['by_channel'].items():
        print(f"   {channel}:")
        print(f"      Transactions: {stats['count']}")
        print(f"      Revenue: ${stats['revenue']:.2f}")

    print(f"\n📦 Sales by Product:")
    # Sort by revenue (descending)
    sorted_products = sorted(
        analytics['by_product'].items(),
        key=lambda x: x[1]['revenue'],
        reverse=True
    )

    for product_id, stats in sorted_products[:5]:  # Top 5
        product_name = catalog.get(product_id, {}).get('name', 'Unknown')
        print(f"   {product_id} ({product_name}):")
        print(f"      Units Sold: {stats['units_sold']}")
        print(f"      Revenue: ${stats['revenue']:.2f}")
````

### Step 5: Generating Daily Report (CSV Export)

````python
# ============================================
# STEP 5: GENERATE DAILY REPORT
# ============================================

def generate_daily_report(analytics, catalog, output_file='daily_report.csv'):
    """
    Generate daily sales report in CSV format.

    Args:
        analytics (dict): Sales analytics data
        catalog (dict): Product catalog
        output_file (str): Output filename
    """
    if not analytics or analytics['total_transactions'] == 0:
        print("❌ No sales data to report")
        return False

    try:
        # Find top product
        top_product_id = max(
            analytics['by_product'].items(),
            key=lambda x: x[1]['revenue']
        )[0]
        top_product_name = catalog.get(top_product_id, {}).get('name', 'Unknown')

        # Prepare report data
        report_date = datetime.now().strftime('%Y-%m-%d')

        # Write summary report
        with open(output_file, 'w', newline='', encoding='utf-8') as file:
            csv_writer = csv.writer(file)

            # Header
            csv_writer.writerow([
                'Date',
                'TotalTransactions',
                'TotalRevenue',
                'AverageTransaction',
                'TopProduct',
                'TopProductRevenue'
            ])

            # Data
            avg_transaction = (analytics['total_revenue'] /
                             analytics['total_transactions'])

            csv_writer.writerow([
                report_date,
                analytics['total_transactions'],
                f"{analytics['total_revenue']:.2f}",
                f"{avg_transaction:.2f}",
                top_product_name,
                f"{analytics['by_product'][top_product_id]['revenue']:.2f}"
            ])

            # Blank row
            csv_writer.writerow([])

            # Channel breakdown
            csv_writer.writerow(['Channel Breakdown'])
            csv_writer.writerow(['Channel', 'Transactions', 'Revenue'])

            for channel, stats in analytics['by_channel'].items():
                csv_writer.writerow([
                    channel,
                    stats['count'],
                    f"{stats['revenue']:.2f}"
                ])

            # Blank row
            csv_writer.writerow([])

            # Product breakdown
            csv_writer.writerow(['Product Breakdown'])
            csv_writer.writerow(['ProductID', 'ProductName', 'UnitsSold', 'Revenue'])

            sorted_products = sorted(
                analytics['by_product'].items(),
                key=lambda x: x[1]['revenue'],
                reverse=True
            )

            for product_id, stats in sorted_products:
                product_name = catalog.get(product_id, {}).get('name', 'Unknown')
                csv_writer.writerow([
                    product_id,
                    product_name,
                    stats['units_sold'],
                    f"{stats['revenue']:.2f}"
                ])

        print(f"✅ Daily report generated: {output_file}")
        return True

    except Exception as e:
        print(f"❌ Error generating report: {e}")
        return False


# Generate report
print("\n" + "=" * 60)
print("GENERATING DAILY REPORT")
print("=" * 60)

generate_daily_report(analytics, catalog)

# Read and display the report
print("\n📄 Report Contents:")
print("=" * 60)

try:
    with open('daily_report.csv', 'r', encoding='utf-8') as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Report file not found")
````

### Step 6: Updating Product Stock in CSV

````python
# ============================================
# STEP 6: UPDATE PRODUCT CATALOG
# ============================================

def save_product_catalog(catalog, filename='products.csv'):
    """
    Save updated product catalog back to CSV file.

    Args:
        catalog (dict): Updated product catalog
        filename (str): Output filename

    Returns:
        bool: Success status
    """
    try:
        with open(filename, 'w', newline='', encoding='utf-8') as file:
            csv_writer = csv.writer(file)

            # Write header
            csv_writer.writerow(['ProductID', 'ProductName', 'Category', 'Price', 'Stock'])

            # Write products (sorted by ProductID)
            for product_id in sorted(catalog.keys()):
                product = catalog[product_id]
                csv_writer.writerow([
                    product_id,
                    product['name'],
                    product['category'],
                    f"{product['price']:.2f}",
                    product['stock']
                ])

        print(f"✅ Updated product catalog saved to {filename}")
        return True

    except Exception as e:
        print(f"❌ Error saving catalog: {e}")
        return False


# Save updated catalog
print("\n" + "=" * 60)
print("UPDATING PRODUCT CATALOG")
print("=" * 60)

save_product_catalog(catalog)

print("\n📦 Updated Stock Levels:")
for product_id in sorted(catalog.keys()):
    product = catalog[product_id]
    stock_status = "✅" if product['stock'] > 10 else "⚠️"
    print(f"{stock_status} {product_id}: {product['name']} - {product['stock']} units")
````

## 12. Common Pitfalls and Solutions

### Pitfall 1: Not Closing Files

**Problem:**
```python
file = open('data.txt', 'w')
file.write('data')
# Forgot to close! Data might not be saved
```

**Solution:**
```python
with open('data.txt', 'w') as file:
    file.write('data')
# Automatically closed, data is saved
```

### Pitfall 2: Overwriting Important Data

**Problem:**
```python
# Opens in 'w' mode - DESTROYS existing content!
with open('important_data.txt', 'w') as file:
    file.write('New data')
```

**Solution:**
```python
# Use 'a' mode to append, or check file existence first
import os

if os.path.exists('important_data.txt'):
    user_confirm = input("File exists. Overwrite? (yes/no): ")
    if user_confirm.lower() != 'yes':
        print("Operation cancelled")
        exit()

with open('important_data.txt', 'w') as file:
    file.write('New data')
```

### Pitfall 3: Ignoring Encoding

**Problem:**
```python
# Might fail with international characters
with open('data.txt', 'r') as file:
    content = file.read()  # UnicodeDecodeError!
```

**Solution:**
```python
# Always specify encoding
with open('data.txt', 'r', encoding='utf-8') as file:
    content = file.read()
```

### Pitfall 4: Memory Issues with Large Files

**Problem:**
```python
# Loads entire 1GB file into memory!
with open('huge_log.txt', 'r') as file:
    content = file.read()  # Memory error!
```

**Solution:**
```python
# Process line by line
with open('huge_log.txt', 'r') as file:
    for line in file:
        process(line)  # Memory efficient
```

### Pitfall 5: Not Handling Errors

**Problem:**
```python
# Crashes if file doesn't exist
with open('data.txt', 'r') as file:
    content = file.read()
```

**Solution:**
```python
try:
    with open('data.txt', 'r') as file:
        content = file.read()
except FileNotFoundError:
    print("File not found, using defaults")
    content = default_content()
```


## 13. Key Takeaways

### Essential Concepts

✅ **Always use `with` statement** for file operations
- Automatically handles closing
- Handles exceptions gracefully
- Industry best practice

✅ **Choose the right file mode**
- `'r'` for reading (safest)
- `'w'` for writing (overwrites!)
- `'a'` for appending (adds to end)

✅ **Use csv module for structured data**
- More robust than manual parsing
- Handles edge cases correctly
- Industry standard

✅ **Handle errors appropriately**
- Files might not exist
- Permissions might be denied
- Data might be corrupted

✅ **Consider performance**
- Process large files line-by-line
- Don't load huge files into memory
- Batch operations when possible
