## **What is a Loop? (The Concept of Repetition)**

"The **'To-Do List' Analogy"**

Imagine you have a simple to-do list:

* Buy milk
* Call mom
* Walk the dog

How would you explain this to a robot?

You would say: 'For each item on my list, do the item.' This 'for each' is a loop!

It's a way to repeat an action for a collection of things.

```python
# === LEVEL 0: WHAT IS A SINGLE LOOP? ===
# We have a list of tasks
tasks = ["Buy milk", "Call mom", "Walk the dog"]

print("My To-Do List:")

# 'task' is a variable that holds one item at a time.
# First, task = "Buy milk"
# Second, task = "Call mom"
# Third, task = "Walk the dog"
for task in tasks:
    print(f"  - I need to {task}")

print("All tasks done!")
```

There are two common ways to loop in Python:

**Looping through a list (The `for...in` format)**:

* **Use when**: You have a list of items and you want to do something with each item.

Example:

`for student in students`: ... (Here, student becomes "Anna", then "Ben", then "Charlie").


**Looping a specific number of times (`The range()` format)**:

* **Use when**: You just want to repeat an action 'X' times (e.g., 5 times, 10 times).

Example:

`for i in range(5)`: ... (This gives you numbers 0, 1, 2, 3, 4). i is just a counter.

Think of it this way: `for...in` is like checking off names on a list. `range()` is like using a tally counter."

# **Understanding Nested Loops - One Step at a Time**

"Think about your school. What's the structure?"

**Analogy**: Departments → Classrooms → Students.

**Key Phrase**: "For every X, do something for every Y."



*   "For every floor in the hotel, check every room."
*   "For every class, check every student."





```python
classes = ["1A", "1B", "2A"]
students = ["Anna", "Ben", "Charlie"]

for class_name in classes:      # Outer Loop
    print(f"📚 Class: {class_name}")
    for student_name in students: # Inner Loop
        print(f"   - Student: {student_name}")
```

**Why Nested Loops?**

A loop allows us to repeat an action multiple times.
A nested loop means a loop inside another loop, used when data or tasks are two-dimensional, layered, multi-grouped, or time-structured.

Real-world examples:

| Scenario              | Outer Loop  | Inner Loop |
| --------------------- | ----------- | ---------- |
| Hotel room inspection | Floors      | Rooms      |
| School attendance     | Classes     | Students   |
| Supermarket audit     | Aisles      | Products   |
| Time analysis         | Months      | Days       |
| Company structure     | Departments | Employees  |


Example phrasing:

**“For every class, check every student.”**

A perfect real-world example of a nested loop is announcing every card in a deck.

How is a deck structured?

* It has 4 Suits (Hearts, Diamonds, Clubs, Spades).

* For each suit, it has 13 Ranks (Ace, 2, 3, ..., King).

This is a nested loop!

* Outer Loop: The 4 Suits.

* Inner Loop: The 13 Ranks.

The logic is: 'For every suit, go through every rank.'

```python
suits = ["Hearts", "Diamonds", "Clubs", "Spades"]
ranks = ["Ace", "2", "3", "4", "5"] # (Shortened for the example)

# Outer Loop: Pick a suit
for suit in suits:
    print(f"--- Announcing all cards for: {suit} ---")

    # Inner Loop: Go through all ranks FOR THAT SUIT
    for rank in ranks:
        print(f"   {rank} of {suit}")
```

**Let's 'Trace' the Code (Be the Computer)**

How does the `(i, j)` code actually run? Let's trace it step-by-step.

`for i in range(1, 4):` (This means i will be 1, 2, 3) `for j in range(1, 4):` (This means j will be 1, 2, 3)


1. Outer loop starts. `i` is now 1.

2. Computer enters the inner loop.

3. Inner loop starts. `j` is now 1.

4. `print(f"({i}, {j})")` ->` Prints (1, 1)`

5. Inner loop repeats. `j` is now 2.

6. `print(f"({i}, {j})")` -> `Prints (1, 2)`

7. Inner loop repeats. `j` is now 3.

8. `print(f"({i}, {j})")` -> `Prints (1, 3)`

9. Inner loop finishes.

10. `print() `is called (the one at the end), moving to a new line.

11. Outer loop repeats. `i` is now 2.

12. Computer re-enters the inner loop.

13. Inner loop starts OVER. `j` is now 1.

14. print(f"({i}, {j})") -> `Prints (2, 1)`

15. Inner loop repeats. `j` is now 2.

16. `print(f"({i}, {j})")` -> `Prints (2, 2)`

...and so on.



**The Key Idea**: The Inner Loop must run to completion for every single step of the Outer Loop."

**Basic Nested Loop Example**

```python
for i in range(1, 4):      # Outer loop
    for j in range(1, 4):  # Inner loop
        print(f"({i}, {j})", end=" ")
    print()                # Moves to next line
```

**Creating a Multiplication Table**

```python
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row * col:2}", end="  ")
    print()
```

**"What if we want ALL tables from 2 to 4?"**

```python
for table in range(2, 5):     # Outer: Tables 2, 3, 4
    print(f"\nTable of {table}:")
    for number in range(1, 6): # Inner: Numbers 1 to 5
        print(f"  {table} x {number} = {table * number}")
```

# **Real-World Business Case: Electronic Equipment Sales Tracking**

**Company**: "TechSphere Electronics"

**Business Structure**:

25 SKUs: iPhone 13, Samsung Galaxy S22, MacBook Pro, etc.

**Sales Channels**:

* Online Stores (Website, Mobile App)

* Chain Electronic Markets (TeknoSA, MediaMarkt, etc.)

* GSM Operator Network (Vodafone, Turk Telekom, etc.)


Each operator has multiple Distributors

Each distributor supplies multiple Sales Points

**Sales Tracking Problem Using Nested Loops**

**Problem Statement:**

"Create a monthly sales tracking system that calculates total sales across all channels, identifies the best-performing product in each channel, and finds the top sales point in the GSM operator network."

```python
import random

# Company Data Setup
skus = [f"SKU_{i:02d}" for i in range(1, 26)]  # SKU_01 to SKU_25
online_stores = ["Website", "Mobile_App", "Amazon_Marketplace"]
chain_markets = ["BestBuy", "MediaMarkt", "ElectroWorld", "Currys"]
gsm_operators = {
    "Vodafone": ["Distributor_A", "Distributor_B"],
    "T-Mobile": ["Distributor_C", "Distributor_D", "Distributor_E"],
    "Orange": ["Distributor_F"]
}
sales_points = {
    "Distributor_A": ["Store_1", "Store_2", "Store_3"],
    "Distributor_B": ["Store_4", "Store_5"],
    "Distributor_C": ["Store_6", "Store_7", "Store_8", "Store_9"],
    "Distributor_D": ["Store_10", "Store_11"],
    "Distributor_E": ["Store_12"],
    "Distributor_F": ["Store_13", "Store_14", "Store_15"]
}

# Initialize sales tracking
monthly_sales = {"Online": {}, "Chain_Markets": {}, "GSM_Network": {}}

print("=== TECHSPHERE ELECTRONICS - MONTHLY SALES REPORT ===\n")

# 1. ONLINE STORES SALES
print("1. ONLINE STORES CHANNEL:")
online_total = 0
for store in online_stores:
    print(f"   📱 {store}:")
    store_sales = 0
    for sku in skus:
        sales = random.randint(50, 200)
        monthly_sales["Online"][(store, sku)] = sales
        store_sales += sales
        print(f"      - {sku}: {sales} units")
    online_total += store_sales
    print(f"   → {store} Total: {store_sales} units\n")
print(f"🟢 ONLINE CHANNEL TOTAL: {online_total} units\n")

# 2. CHAIN MARKETS SALES
print("2. CHAIN ELECTRONIC MARKETS:")
chain_total = 0
for market in chain_markets:
    print(f"   🏪 {market}:")
    market_sales = 0
    for sku in skus:
        sales = random.randint(30, 150)
        monthly_sales["Chain_Markets"][(market, sku)] = sales
        market_sales += sales
        print(f"      - {sku}: {sales} units")
    chain_total += market_sales
    print(f"   → {market} Total: {market_sales} units\n")
print(f"🟢 CHAIN MARKETS TOTAL: {chain_total} units\n")

# 3. GSM OPERATOR NETWORK SALES (TRIPLE NESTED LOOP!)
print("3. GSM OPERATOR NETWORK:")
gsm_total = 0
sales_point_performance = {}  # To find the best sales point

for operator, distributors in gsm_operators.items():
    print(f"   📞 {operator}:")
    operator_total = 0

    for distributor in distributors:
        print(f"      🚚 {distributor}:")
        distributor_total = 0

        for sales_point in sales_points[distributor]:
            print(f"         🏬 {sales_point}:")
            point_sales = 0

            for sku in skus:
                sales = random.randint(5, 50)  # Lower volume per store
                monthly_sales["GSM_Network"][(operator, distributor, sales_point, sku)] = sales
                point_sales += sales
                print(f"            - {sku}: {sales} units")

            sales_point_performance[sales_point] = point_sales
            distributor_total += point_sales
            print(f"         → {sales_point} Total: {point_sales} units")

        operator_total += distributor_total
        print(f"      → {distributor} Total: {distributor_total} units")

    gsm_total += operator_total
    print(f"   → {operator} Total: {operator_total} units\n")

print(f"🟢 GSM NETWORK TOTAL: {gsm_total} units\n")

# FINAL SUMMARY
grand_total = online_total + chain_total + gsm_total
print("=" * 50)
print("📊 FINAL MONTHLY SUMMARY:")
print(f"Online Stores: {online_total} units")
print(f"Chain Markets: {chain_total} units")
print(f"GSM Network: {gsm_total} units")
print(f"🔥 GRAND TOTAL: {grand_total} units")
print("=" * 50)

# BONUS ANALYSIS
print("\n🏆 PERFORMANCE ANALYSIS:")

# Find best performing sales point in GSM network
best_point = max(sales_point_performance, key=sales_point_performance.get)
print(f"⭐ Best Performing Sales Point: {best_point} ({sales_point_performance[best_point]} units)")

# Find best selling SKU overall (simplified)
print(f"📈 Total Products Sold: {grand_total} across {len(skus)} SKUs")
```
