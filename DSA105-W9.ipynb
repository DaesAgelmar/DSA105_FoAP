# **Lists and Single Loops**

In real life, we constantly work with collections:

* A shopping list contains multiple items
* A classroom roster contains multiple student names
* A bank account has multiple transactions
* A store inventory has multiple products

**The Problem Without Lists:**

student1 = "Alice"

student2 = "Bob"

student3 = "Charlie"

student4 = "Diana"

student5 = "Eve"

 What if we have 100 students?
 What if we need to add a new student?
 This approach doesn't scale!

We have discussed basic concepts of lists in W5 files.


**Single Loops - The "For Each" Concept**

```python
# ============================================
# PROBLEM: Repetitive, Not Scalable
# ============================================

tasks = ["Buy milk", "Call mom", "Walk dog", "Study Python", "Exercise"]

# Printing each task manually
print("Task:", tasks[0])
print("Task:", tasks[1])
print("Task:", tasks[2])
print("Task:", tasks[3])
print("Task:", tasks[4])

# What if we have 100 tasks?
# What if we add or remove tasks?
# This code is FRAGILE and TEDIOUS!
```

**The Solution: For Loop**

A loop is a way to say: **"For EACH item in my list, do something"**

Real-life analogy:

* Teacher: "For each student in the class, mark attendance"
* Cashier: "For each item in the cart, scan the barcode"
* Manager: "For each employee, calculate monthly salary"

```python
# Basic for Loop#
# ============================================
# EXAMPLE : Your First Loop
# ============================================

tasks = ["Buy milk", "Call mom", "Walk dog"]

# The loop structure:
# for [item_variable] in [list_name]:
#     do something with item_variable

for task in tasks:
    print(f"✓ {task}")

# How it works:
# Step 1: task = "Buy milk" → print "✓ Buy milk"
# Step 2: task = "Call mom" → print "✓ Call mom"
# Step 3: task = "Walk dog" → print "✓ Walk dog"
# Step 4: No more items → loop ends
```

**Loop with Calculations**

```python
# ============================================
# EXAMPLE : Processing Numeric Data
# ============================================

prices = [10, 25, 15, 30, 20]

print("Price List:")
print("=" * 30)

for price in prices:
    print(f"Item price: ${price}")

# Calculate total
total = 0
for price in prices:
    total = total + price

print("=" * 30)
print(f"Total: ${total}")
```

**Real-World Business Example:**

```python
# ============================================
# BUSINESS EXAMPLE: Daily Sales Report
# ============================================

daily_sales = [150, 200, 175, 300, 250]
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

print("📊 WEEKLY SALES REPORT")
print("=" * 40)

# Note: We have two lists of same length
# We'll learn to use both together later

sales_total = 0
for sale in daily_sales:
    sales_total = sales_total + sale

average_sale = sales_total / len(daily_sales)

print(f"Total Weekly Sales: ${sales_total}")
print(f"Average Daily Sale: ${average_sale:.2f}")

# Find best sales day
max_sale = 0
for sale in daily_sales:
    if sale > max_sale:
        max_sale = sale

print(f"Best Sales Day: ${max_sale}")
```

**Loop Variable Naming**

* List name: plural (students, prices, fruits)
* Loop variable: singular (student, price, fruit)

 This makes code self-documenting!

```python
# ============================================
# EXAMPLE: Good vs Bad Variable Names
# ============================================

# ❌ BAD: Unclear variable names
for x in ["Apple", "Banana", "Orange"]:
    print(x)

# ❌ BAD: Generic names
for item in ["Apple", "Banana", "Orange"]:
    print(item)

# ✅ GOOD: Descriptive names
for fruit in ["Apple", "Banana", "Orange"]:
    print(f"I like {fruit}")

# ✅ GOOD: Context-specific names
students = ["Alice", "Bob", "Charlie"]
for student in students:
    print(f"Student name: {student}")

prices = [10, 20, 30]
for price in prices:
    print(f"Price: ${price}")
```

**Loops with `range()` - Counting and Repetition**

`range()` is a number generator. It creates a sequence of numbers that we can loop through.

When to use `range()`:

* When you need to repeat something N times
* When you need to count from A to B
* When you need numbers for calculations or positions

```python
# ============================================
# EXAMPLE: Understanding range()
# ============================================

# range(n) generates: 0, 1, 2, ..., n-1
print("range(5) generates:")
for i in range(5):
    print(i, end=" ")  # Output: 0 1 2 3 4
print()

# Why does it stop at 4, not 5?
# Python counts from 0 and EXCLUDES the stop value
# This is consistent with list indexing!

# range(start, stop) generates: start, start+1, ..., stop-1
print("\nrange(1, 6) generates:")
for i in range(1, 6):
    print(i, end=" ")  # Output: 1 2 3 4 5
print()

# range(start, stop, step)
print("\nrange(0, 10, 2) generates:")
for i in range(0, 10, 2):
    print(i, end=" ")  # Output: 0 2 4 6 8
print()
```

**Practical `range()` Applications**

```python
# ============================================
# EXAMPLE : Real-World range() Usage
# ============================================

# Application 1: Customer Greetings
print("Welcome Messages:")
for customer_number in range(1, 6):
    print(f"Welcome, Customer #{customer_number}!")

# Application 2: Sequential Processing
print("\nProcessing Orders:")
for order_id in range(1001, 1006):
    print(f"Order #{order_id} - Status: Shipped")

# Application 3: Countdown Timer
print("\nCountdown:")
for seconds in range(5, 0, -1):
    print(f"{seconds}...")
print("Go!")

# Application 4: Multiplication Table
print("\nMultiplication Table for 7:")
for i in range(1, 11):
    result = 7 * i
    print(f"7 × {i} = {result}")
```

** Combining `range()` with Lists**

```python
# ============================================
# EXAMPLE : Using range() to Access List by Index
# ============================================

products = ["Laptop", "Mouse", "Keyboard", "Monitor"]

# Method 1: Direct iteration (PREFERRED)
print("Method 1: Direct iteration")
for product in products:
    print(f"  - {product}")

# Method 2: Using range() and index (when you need position)
print("\nMethod 2: Using index")
for i in range(len(products)):
    print(f"  Position {i}: {products[i]}")

# When to use Method 2?
# - When you need the position number
# - When you need to modify the list
# - When you need to compare with other lists

# Real-world example: Price list with numbering
products = ["Laptop", "Mouse", "Keyboard"]
prices = [1000, 25, 75]

print("\nProduct Catalog:")
for i in range(len(products)):
    item_number = i + 1  # Display as 1, 2, 3 instead of 0, 1, 2
    print(f"{item_number}. {products[i]} - ${prices[i]}")
```

## **Common Patterns and Best Practices**

**Pattern 1: Accumulator Pattern**

```python
# ============================================
# PATTERN 1: Accumulator (Building Up Values)
# ============================================

# Pattern structure:
# 1. Initialize accumulator variable
# 2. Loop through data
# 3. Update accumulator each iteration

# Example: Sum of numbers
numbers = [10, 20, 30, 40, 50]
total = 0  # Step 1: Initialize

for number in numbers:  # Step 2: Loop
    total = total + number  # Step 3: Accumulate

print(f"Total: {total}")

# Example: Building a string
words = ["Hello", "Python", "World"]
sentence = ""  # Initialize

for word in words:
    sentence = sentence + word + " "  # Accumulate

print(f"Sentence: {sentence.strip()}")
```

**Pattern 2: Counter Pattern**

```python
# ============================================
# PATTERN 2: Counter (Counting Occurrences)
# ============================================

# Pattern structure:
# 1. Initialize counter(s)
# 2. Loop through data
# 3. Increment counter when condition met

# Example: Count positive vs negative
numbers = [10, -5, 20, -15, 30, -10, 40]
positive_count = 0  # Initialize counters
negative_count = 0

for number in numbers:
    if number > 0:
        positive_count = positive_count + 1
    elif number < 0:
        negative_count = negative_count + 1

print(f"Positive numbers: {positive_count}")
print(f"Negative numbers: {negative_count}")
```

**Pattern 3: Filter Pattern**

```python
# ============================================
# PATTERN 3: Filter (Selecting Items)
# ============================================

# Pattern structure:
# 1. Create empty result list
# 2. Loop through source data
# 3. Add items that meet criteria

# Example: Select expensive products
all_prices = [15, 150, 25, 200, 10, 300]
expensive = []  # Initialize result list

for price in all_prices:
    if price > 100:  # Condition
        expensive.append(price)  # Add to result

print(f"Expensive items: {expensive}")

# Example: Select students with high grades
students = ["Alice", "Bob", "Charlie", "Diana"]
grades = [85, 92, 78, 95]
honor_roll = []

for i in range(len(students)):
    if grades[i] >= 90:
        honor_roll.append(students[i])

print(f"Honor Roll: {honor_roll}")
```

**Pattern 4: Transform Pattern**

```python
# ============================================
# PATTERN 4: Transform (Modifying Data)
# ============================================

# Pattern structure:
# 1. Create new list for results
# 2. Loop through source data
# 3. Apply transformation to each item

# Example: Apply discount to prices
original_prices = [100, 200, 150, 300]
discounted_prices = []

discount_rate = 0.20  # 20% discount

for price in original_prices:
    new_price = price * (1 - discount_rate)
    discounted_prices.append(new_price)

print("Original prices:", original_prices)
print("Discounted prices:", discounted_prices)

# Example: Convert temperatures Celsius to Fahrenheit
celsius_temps = [0, 10, 20, 30, 40]
fahrenheit_temps = []

for celsius in celsius_temps:
    fahrenheit = (celsius * 9/5) + 32
    fahrenheit_temps.append(fahrenheit)

print(f"Celsius: {celsius_temps}")
print(f"Fahrenheit: {fahrenheit_temps}")
```
