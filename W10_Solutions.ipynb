## 🎯 Task 1: Calculate Channel Percentages

### Complete Working Code

```python
import random

# ============================================
# DATA GENERATION USING DICTIONARIES
# ============================================

print("=" * 70)
print("TASK 1: CHANNEL PERCENTAGE ANALYSIS")
print("=" * 70)

# Initialize sales tracking dictionary
monthly_sales = {
    "Online": {},
    "Chain_Markets": {},
    "GSM_Network": {}
}

# Channel totals dictionary
channel_totals = {
    "Online": 0,
    "Chain_Markets": 0,
    "GSM_Network": 0
}

print("\n📊 Generating sales data...\n")

# ============================================
# 1. ONLINE STORES SALES (Using FOR loop)
# ============================================
print("1️⃣  ONLINE STORES:")

for store in ONLINE_STORES:
    print(f"\n   📱 {store}:")
    store_total = 0

    for sku_code, sku_name, sku_price in PRODUCTS:
        # Generate random sales
        units_sold = random.randint(50, 200)
        revenue = units_sold * sku_price

        # Store in dictionary
        monthly_sales["Online"][(store, sku_code)] = {
            "units": units_sold,
            "revenue": revenue,
            "product_name": sku_name
        }

        store_total += units_sold

    print(f"      Total: {store_total:,} units")
    channel_totals["Online"] += store_total

print(f"\n   ✅ Online Channel Total: {channel_totals['Online']:,} units")

# ============================================
# 2. CHAIN MARKETS SALES (Using FOR loop)
# ============================================
print("\n2️⃣  CHAIN MARKETS:")

for market in CHAIN_MARKETS:
    print(f"\n   🏪 {market}:")
    market_total = 0

    for sku_code, sku_name, sku_price in PRODUCTS:
        units_sold = random.randint(30, 150)
        revenue = units_sold * sku_price

        monthly_sales["Chain_Markets"][(market, sku_code)] = {
            "units": units_sold,
            "revenue": revenue,
            "product_name": sku_name
        }

        market_total += units_sold

    print(f"      Total: {market_total:,} units")
    channel_totals["Chain_Markets"] += market_total

print(f"\n   ✅ Chain Markets Total: {channel_totals['Chain_Markets']:,} units")

# ============================================
# 3. GSM NETWORK SALES (Using NESTED FOR loops)
# ============================================
print("\n3️⃣  GSM NETWORK:")

for operator, distributors in GSM_OPERATORS.items():
    print(f"\n   📞 {operator}:")
    operator_total = 0

    for distributor in distributors:
        print(f"      🚚 {distributor}:")
        distributor_total = 0

        for sales_point in SALES_POINTS[distributor]:
            point_total = 0

            for sku_code, sku_name, sku_price in PRODUCTS:
                units_sold = random.randint(5, 50)
                revenue = units_sold * sku_price

                monthly_sales["GSM_Network"][(operator, distributor, sales_point, sku_code)] = {
                    "units": units_sold,
                    "revenue": revenue,
                    "product_name": sku_name
                }

                point_total += units_sold

            print(f"         🏬 {sales_point}: {point_total} units")
            distributor_total += point_total

        print(f"      → {distributor} Total: {distributor_total:,} units")
        operator_total += distributor_total

    print(f"   → {operator} Total: {operator_total:,} units")
    channel_totals["GSM_Network"] += operator_total

print(f"\n   ✅ GSM Network Total: {channel_totals['GSM_Network']:,} units")

# ============================================
# CALCULATE GRAND TOTAL
# ============================================
grand_total = sum(channel_totals.values())

print("\n" + "=" * 70)
print("📈 MONTHLY SALES SUMMARY")
print("=" * 70)
print(f"Online Stores:    {channel_totals['Online']:>10,} units")
print(f"Chain Markets:    {channel_totals['Chain_Markets']:>10,} units")
print(f"GSM Network:      {channel_totals['GSM_Network']:>10,} units")
print("-" * 70)
print(f"GRAND TOTAL:      {grand_total:>10,} units")
print("=" * 70)

# ============================================
# TASK 1 SOLUTION: CALCULATE PERCENTAGES
# ============================================

print("\n" + "=" * 70)
print("📊 TASK 1 SOLUTION: CHANNEL DISTRIBUTION ANALYSIS")
print("=" * 70)

# Calculate percentages for each channel
channel_percentages = {}

# Using WHILE loop to iterate through channels
channels_list = list(channel_totals.keys())
channel_index = 0

while channel_index < len(channels_list):
    channel_name = channels_list[channel_index]
    channel_sales = channel_totals[channel_name]

    # Calculate percentage
    percentage = (channel_sales / grand_total) * 100
    channel_percentages[channel_name] = percentage

    channel_index += 1

# Display results with visual representation
print("\n📊 Channel Distribution:")
print("-" * 70)

for channel_name, percentage in channel_percentages.items():
    units = channel_totals[channel_name]

    # Create visual bar (each █ represents 2%)
    bar_length = int(percentage // 2)
    visual_bar = "█" * bar_length

    # Determine emoji based on performance
    if percentage >= 40:
        emoji = "🟢"
    elif percentage >= 30:
        emoji = "🟡"
    else:
        emoji = "🔴"

    print(f"\n{emoji} {channel_name:15}")
    print(f"   Percentage:  {percentage:>6.2f}%")
    print(f"   Units Sold:  {units:>10,}")
    print(f"   Visual:      {visual_bar}")

# ============================================
# IDENTIFY DOMINANT CHANNEL (Using WHILE loop)
# ============================================

print("\n" + "=" * 70)
print("🏆 DOMINANT CHANNEL ANALYSIS")
print("=" * 70)

# Find max using while loop
max_percentage = 0
dominant_channel = ""
channel_index = 0

while channel_index < len(channels_list):
    channel_name = channels_list[channel_index]
    percentage = channel_percentages[channel_name]

    if percentage > max_percentage:
        max_percentage = percentage
        dominant_channel = channel_name

    channel_index += 1

print(f"\n🥇 Dominant Channel: {dominant_channel}")
print(f"   Market Share: {max_percentage:.2f}%")
print(f"   Total Units: {channel_totals[dominant_channel]:,}")

# Calculate dominance level
if max_percentage >= 45:
    dominance_level = "VERY STRONG"
elif max_percentage >= 35:
    dominance_level = "STRONG"
else:
    dominance_level = "BALANCED"

print(f"   Dominance: {dominance_level}")

print("\n" + "=" * 70)
print("✅ TASK 1 COMPLETE")
print("=" * 70)
```

## 🏆 Task 2: Find Best-Selling SKU in Each Channel

### Complete Working Code

```python
import random

# [Use the same data generation from Task 1]
# [monthly_sales dictionary is already populated]

print("\n" + "=" * 70)
print("TASK 2: BEST-SELLING SKU ANALYSIS")
print("=" * 70)

# ============================================
# AGGREGATE SALES BY SKU FOR EACH CHANNEL
# ============================================

# Initialize dictionaries to store SKU totals
sku_sales_online = {}
sku_sales_chain = {}
sku_sales_gsm = {}

print("\n📊 Analyzing sales by product...\n")

# ============================================
# ONLINE STORES - Aggregate using WHILE loop
# ============================================
print("1️⃣  Analyzing Online Stores...")

# Convert dictionary items to list for while loop
online_items = list(monthly_sales["Online"].items())
item_index = 0

while item_index < len(online_items):
    (store, sku_code), sale_data = online_items[item_index]

    # Initialize if not exists
    if sku_code not in sku_sales_online:
        sku_sales_online[sku_code] = {
            "total_units": 0,
            "total_revenue": 0,
            "product_name": sale_data["product_name"]
        }

    # Accumulate sales
    sku_sales_online[sku_code]["total_units"] += sale_data["units"]
    sku_sales_online[sku_code]["total_revenue"] += sale_data["revenue"]

    item_index += 1

print(f"   ✅ Processed {len(sku_sales_online)} products")

# ============================================
# CHAIN MARKETS - Aggregate using FOR loop
# ============================================
print("2️⃣  Analyzing Chain Markets...")

for (market, sku_code), sale_data in monthly_sales["Chain_Markets"].items():
    if sku_code not in sku_sales_chain:
        sku_sales_chain[sku_code] = {
            "total_units": 0,
            "total_revenue": 0,
            "product_name": sale_data["product_name"]
        }

    sku_sales_chain[sku_code]["total_units"] += sale_data["units"]
    sku_sales_chain[sku_code]["total_revenue"] += sale_data["revenue"]

print(f"   ✅ Processed {len(sku_sales_chain)} products")

# ============================================
# GSM NETWORK - Aggregate using NESTED loops
# ============================================
print("3️⃣  Analyzing GSM Network...")

for (operator, distributor, point, sku_code), sale_data in monthly_sales["GSM_Network"].items():
    if sku_code not in sku_sales_gsm:
        sku_sales_gsm[sku_code] = {
            "total_units": 0,
            "total_revenue": 0,
            "product_name": sale_data["product_name"]
        }

    sku_sales_gsm[sku_code]["total_units"] += sale_data["units"]
    sku_sales_gsm[sku_code]["total_revenue"] += sale_data["revenue"]

print(f"   ✅ Processed {len(sku_sales_gsm)} products")

# ============================================
# FIND BEST-SELLING SKU IN EACH CHANNEL
# ============================================

print("\n" + "=" * 70)
print("🏆 BEST-SELLING PRODUCTS BY CHANNEL")
print("=" * 70)

# Function to find best seller using WHILE loop
def find_best_seller_while(sku_dict, channel_name):
    """Find best-selling SKU using while loop"""

    sku_list = list(sku_dict.keys())

    if len(sku_list) == 0:
        return None, None

    # Find max using while loop
    best_sku = sku_list[0]
    max_units = sku_dict[best_sku]["total_units"]

    index = 1
    while index < len(sku_list):
        current_sku = sku_list[index]
        current_units = sku_dict[current_sku]["total_units"]

        if current_units > max_units:
            max_units = current_units
            best_sku = current_sku

        index += 1

    return best_sku, sku_dict[best_sku]

# ============================================
# ONLINE STORES - Best Seller
# ============================================
print("\n📱 ONLINE STORES:")
print("-" * 70)

best_online_sku, best_online_data = find_best_seller_while(sku_sales_online, "Online")

print(f"\n🥇 Best-Selling Product:")
print(f"   SKU:         {best_online_sku}")
print(f"   Name:        {best_online_data['product_name']}")
print(f"   Units Sold:  {best_online_data['total_units']:,}")
print(f"   Revenue:     ${best_online_data['total_revenue']:,.2f}")
print(f"   Market Share: {(best_online_data['total_units']/channel_totals['Online'])*100:.2f}%")

# Show Top 5 using sorted list
print("\n   📊 Top 5 Products:")
sorted_online = sorted(sku_sales_online.items(),
                       key=lambda x: x[1]["total_units"],
                       reverse=True)

rank = 1
for sku, data in sorted_online[:5]:
    print(f"   {rank}. {sku} ({data['product_name'][:20]:20}): {data['total_units']:>5,} units")
    rank += 1

# ============================================
# CHAIN MARKETS - Best Seller
# ============================================
print("\n🏪 CHAIN MARKETS:")
print("-" * 70)

best_chain_sku, best_chain_data = find_best_seller_while(sku_sales_chain, "Chain")

print(f"\n🥇 Best-Selling Product:")
print(f"   SKU:         {best_chain_sku}")
print(f"   Name:        {best_chain_data['product_name']}")
print(f"   Units Sold:  {best_chain_data['total_units']:,}")
print(f"   Revenue:     ${best_chain_data['total_revenue']:,.2f}")
print(f"   Market Share: {(best_chain_data['total_units']/channel_totals['Chain_Markets'])*100:.2f}%")

print("\n   📊 Top 5 Products:")
sorted_chain = sorted(sku_sales_chain.items(),
                     key=lambda x: x[1]["total_units"],
                     reverse=True)

rank = 1
for sku, data in sorted_chain[:5]:
    print(f"   {rank}. {sku} ({data['product_name'][:20]:20}): {data['total_units']:>5,} units")
    rank += 1

# ============================================
# GSM NETWORK - Best Seller
# ============================================
print("\n📞 GSM NETWORK:")
print("-" * 70)

best_gsm_sku, best_gsm_data = find_best_seller_while(sku_sales_gsm, "GSM")

print(f"\n🥇 Best-Selling Product:")
print(f"   SKU:         {best_gsm_sku}")
print(f"   Name:        {best_gsm_data['product_name']}")
print(f"   Units Sold:  {best_gsm_data['total_units']:,}")
print(f"   Revenue:     ${best_gsm_data['total_revenue']:,.2f}")
print(f"   Market Share: {(best_gsm_data['total_units']/channel_totals['GSM_Network'])*100:.2f}%")

print("\n   📊 Top 5 Products:")
sorted_gsm = sorted(sku_sales_gsm.items(),
                   key=lambda x: x[1]["total_units"],
                   reverse=True)

rank = 1
for sku, data in sorted_gsm[:5]:
    print(f"   {rank}. {sku} ({data['product_name'][:20]:20}): {data['total_units']:>5,} units")
    rank += 1

# ============================================
# OVERALL BEST SELLER (All Channels Combined)
# ============================================
print("\n" + "=" * 70)
print("🌟 OVERALL CHAMPION PRODUCT")
print("=" * 70)

# Combine all channels
overall_sku_sales = {}

# Combine using WHILE loop
for sku_code, sku_name, sku_price in PRODUCTS:
    overall_sku_sales[sku_code] = {
        "product_name": sku_name,
        "total_units": 0,
        "online_units": 0,
        "chain_units": 0,
        "gsm_units": 0
    }

    # Add from online
    if sku_code in sku_sales_online:
        overall_sku_sales[sku_code]["online_units"] = sku_sales_online[sku_code]["total_units"]
        overall_sku_sales[sku_code]["total_units"] += sku_sales_online[sku_code]["total_units"]

    # Add from chain
    if sku_code in sku_sales_chain:
        overall_sku_sales[sku_code]["chain_units"] = sku_sales_chain[sku_code]["total_units"]
        overall_sku_sales[sku_code]["total_units"] += sku_sales_chain[sku_code]["total_units"]

    # Add from GSM
    if sku_code in sku_sales_gsm:
        overall_sku_sales[sku_code]["gsm_units"] = sku_sales_gsm[sku_code]["total_units"]
        overall_sku_sales[sku_code]["total_units"] += sku_sales_gsm[sku_code]["total_units"]

# Find overall best using while loop
best_overall_sku, best_overall_data = find_best_seller_while(
    {k: {"total_units": v["total_units"], "product_name": v["product_name"]}
     for k, v in overall_sku_sales.items()},
    "Overall"
)

print(f"\n🏆 CHAMPION PRODUCT: {best_overall_data['product_name']}")
print(f"   SKU: {best_overall_sku}")
print(f"   Total Units: {best_overall_data['total_units']:,}")
print(f"   Market Share: {(best_overall_data['total_units']/grand_total)*100:.2f}%")

print(f"\n   📊 Breakdown by Channel:")
overall_data = overall_sku_sales[best_overall_sku]
print(f"   - Online Stores:  {overall_data['online_units']:>6,} units")
print(f"   - Chain Markets:  {overall_data['chain_units']:>6,} units")
print(f"   - GSM Network:    {overall_data['gsm_units']:>6,} units")

print("\n" + "=" * 70)
print("✅ TASK 2 COMPLETE")
print("=" * 70)
```

## 📅 Task 3: Monthly Comparison

### Complete Working Code

```python
import random

print("\n" + "=" * 70)
print("TASK 3: MULTI-MONTH PERFORMANCE TRACKING")
print("=" * 70)

# ============================================
# DATA STRUCTURE FOR MULTIPLE MONTHS
# ============================================

# Months tuple (immutable)
MONTHS = ("January", "February", "March")

# Dictionary to store quarterly data
quarterly_data = {}

# ============================================
# GENERATE 3 MONTHS OF DATA USING WHILE LOOP
# ============================================

month_index = 0

while month_index < len(MONTHS):
    current_month = MONTHS[month_index]

    print(f"\n{'='*70}")
    print(f"📆 GENERATING DATA FOR: {current_month.upper()}")
    print(f"{'='*70}")

    # Initialize month data
    quarterly_data[current_month] = {
        "sales": {},
        "totals": {
            "Online": 0,
            "Chain_Markets": 0,
            "GSM_Network": 0,
            "Grand_Total": 0
        }
    }

    month_sales = quarterly_data[current_month]["sales"]
    month_totals = quarterly_data[current_month]["totals"]

    # Growth factor (sales increase month-over-month)
    growth_factor = 1.0 + (month_index * 0.10)

    # ============================================
    # ONLINE STORES
    # ============================================
    print(f"\n📱 Online Stores:")

    for store in ONLINE_STORES:
        store_total = 0

        for sku_code, sku_name, sku_price in PRODUCTS:
            base_sales = random.randint(50, 200)
            units = int(base_sales * growth_factor)

            key = ("Online", store, sku_code)
            month_sales[key] = {
                "units": units,
                "revenue": units * sku_price,
                "product_name": sku_name
            }

            store_total += units

        print(f"   - {store:25}: {store_total:>6,} units")
        month_totals["Online"] += store_total

    print(f"   {'→ Total':25}: {month_totals['Online']:>6,} units")

    # ============================================
    # CHAIN MARKETS
    # ============================================
    print(f"\n🏪 Chain Markets:")

    for market in CHAIN_MARKETS:
        market_total = 0

        for sku_code, sku_name, sku_price in PRODUCTS:
            base_sales = random.randint(30, 150)
            units = int(base_sales * growth_factor)

            key = ("Chain", market, sku_code)
            month_sales[key] = {
                "units": units,
                "revenue": units * sku_price,
                "product_name": sku_name
            }

            market_total += units

        print(f"   - {market:25}: {market_total:>6,} units")
        month_totals["Chain_Markets"] += market_total

    print(f"   {'→ Total':25}: {month_totals['Chain_Markets']:>6,} units")

    # ============================================
    # GSM NETWORK
    # ============================================
    print(f"\n📞 GSM Network:")

    for operator, distributors in GSM_OPERATORS.items():
        operator_total = 0

        for distributor in distributors:
            for point in SALES_POINTS[distributor]:
                point_total = 0

                for sku_code, sku_name, sku_price in PRODUCTS:
                    base_sales = random.randint(5, 50)
                    units = int(base_sales * growth_factor)

                    key = ("GSM", operator, distributor, point, sku_code)
                    month_sales[key] = {
                        "units": units,
                        "revenue": units * sku_price,
                        "product_name": sku_name
                    }

                    point_total += units

                operator_total += point_total

        print(f"   - {operator:25}: {operator_total:>6,} units")
        month_totals["GSM_Network"] += operator_total

    print(f"   {'→ Total':25}: {month_totals['GSM_Network']:>6,} units")

    # Calculate grand total for month
    month_totals["Grand_Total"] = (
        month_totals["Online"] +
        month_totals["Chain_Markets"] +
        month_totals["GSM_Network"]
    )

    print(f"\n   🔥 {current_month} GRAND TOTAL: {month_totals['Grand_Total']:,} units")

    month_index += 1

# ============================================
# COMPARATIVE ANALYSIS TABLE
# ============================================

print("\n" + "=" * 70)
print("📊 QUARTERLY COMPARATIVE ANALYSIS")
print("=" * 70)

# Create table header
print(f"\n{'Channel':<20} {'January':>12} {'February':>12} {'March':>12} {'Trend':>12}")
print("-" * 70)

# Analyze each channel
channels = ["Online", "Chain_Markets", "GSM_Network"]
channel_names = {
    "Online": "Online Stores",
    "Chain_Markets": "Chain Markets",
    "GSM_Network": "GSM Network"
}

for channel in channels:
    # Get data for all three months
    jan_units = quarterly_data["January"]["totals"][channel]
    feb_units = quarterly_data["February"]["totals"][channel]
    mar_units = quarterly_data["March"]["totals"][channel]

    # Determine trend using while loop comparison
    if jan_units < feb_units and feb_units < mar_units:
        trend = "📈 UP"
    elif jan_units > feb_units and feb_units > mar_units:
        trend = "📉 DOWN"
    else:
        trend = "➡️ MIXED"

    print(f"{channel_names[channel]:<20} {jan_units:>12,} {feb_units:>12,} {mar_units:>12,} {trend:>12}")

# Print totals row
print("-" * 70)
jan_total = quarterly_data["January"]["totals"]["Grand_Total"]
feb_total = quarterly_data["February"]["totals"]["Grand_Total"]
mar_total = quarterly_data["March"]["totals"]["Grand_Total"]

print(f"{'GRAND TOTAL':<20} {jan_total:>12,} {feb_total:>12,} {mar_total:>12,}")
print("=" * 70)

# ============================================
# GROWTH RATE ANALYSIS
# ============================================

print("\n" + "=" * 70)
print("📈 GROWTH RATE ANALYSIS")
print("=" * 70)

print("\n🔍 Channel Growth (January → March):")
print("-" * 70)

# Calculate growth for each channel using while loop
channel_index = 0

while channel_index < len(channels):
    channel = channels[channel_index]
    channel_name = channel_names[channel]

    jan_value = quarterly_data["January"]["totals"][channel]
    mar_value = quarterly_data["March"]["totals"][channel]

    # Calculate growth rate
    growth_rate = ((mar_value - jan_value) / jan_value) * 100

    # Determine status
    if growth_rate > 0:
        status = "🟢 GROWING"
        direction = "increased by"
    elif growth_rate < 0:
        status = "🔴 DECLINING"
        direction = "decreased by"
    else:
        status = "🟡 FLAT"
        direction = "remained at"

    print(f"\n{status} {channel_name}:")
    print(f"   January:     {jan_value:>10,} units")
    print(f"   March:       {mar_value:>10,} units")
    print(f"   Growth:      {growth_rate:>9.2f}%")
    print(f"   Status:      Sales {direction} {abs(growth_rate):.1f}%")

    channel_index += 1

# Overall company growth
overall_growth = ((mar_total - jan_total) / jan_total) * 100
print(f"\n{'='*70}")
print(f"🎯 OVERALL COMPANY GROWTH: {overall_growth:+.2f}%")
print(f"{'='*70}")

# ============================================
# MONTH-OVER-MONTH CHANGES
# ============================================

print("\n" + "=" * 70)
print("📊 MONTH-OVER-MONTH CHANGES")
print("=" * 70)

# January → February
print("\n📅 January → February:")
jan_feb_change = feb_total - jan_total
jan_feb_percent = (jan_feb_change / jan_total) * 100

print(f"   Absolute Change: {jan_feb_change:+,} units")
print(f"   Percentage:      {jan_feb_percent:+.2f}%")

if jan_feb_percent > 0:
    print(f"   Status:          ✅ Growth")
else:
    print(f"   Status:          ⚠️  Decline")

# February → March
print("\n📅 February → March:")
feb_mar_change = mar_total - feb_total
feb_mar_percent = (feb_mar_change / feb_total) * 100

print(f"   Absolute Change: {feb_mar_change:+,} units")
print(f"   Percentage:      {feb_mar_percent:+.2f}%")

if feb_mar_percent > 0:
    print(f"   Status:          ✅ Growth")
else:
    print(f"   Status:          ⚠️  Decline")

# ============================================
# APRIL PROJECTION (Simple Forecast)
# ============================================

print("\n" + "=" * 70)
print("🔮 APRIL SALES PROJECTION")
print("=" * 70)

# Calculate average monthly growth
total_growth = mar_total - jan_total
avg_monthly_growth = total_growth / 2

# Project April
april_projection = mar_total + avg_monthly_growth

print(f"\n📊 Forecast Methodology: Linear Trend")
print(f"   Average Monthly Growth: {avg_monthly_growth:+,.0f} units")
print(f"   March Actual:           {mar_total:>10,} units")
print(f"   April Projection:       {april_projection:>10,.0f} units")

# Confidence level based on consistency
if abs(jan_feb_percent - feb_mar_percent) < 2:
    confidence = "HIGH"
    emoji = "🟢"
else:
    confidence = "MODERATE"
    emoji = "🟡"

print(f"   Confidence Level:       {emoji} {confidence}")

# ============================================
# BEST & WORST PERFORMING CHANNELS
# ============================================

print("\n" + "=" * 70)
print("🏆 QUARTERLY PERFORMANCE RANKING")
print("=" * 70)

# Calculate quarterly totals
channel_quarterly = {}

for channel in channels:
    total = (
        quarterly_data["January"]["totals"][channel] +
        quarterly_data["February"]["totals"][channel] +
        quarterly_data["March"]["totals"][channel]
    )
    channel_quarterly[channel] = total

# Find best and worst using while loop
best_channel = None
worst_channel = None
max_sales = 0
min_sales = float('inf')

channel_index = 0
while channel_index < len(channels):
    channel = channels[channel_index]
    total = channel_quarterly[channel]

    if total > max_sales:
        max_sales = total
        best_channel = channel

    if total < min_sales:
        min_sales = total
        worst_channel = channel

    channel_index += 1

print(f"\n🥇 Best Performing Channel: {channel_names[best_channel]}")
print(f"   Quarterly Total:  {max_sales:>10,} units")
print(f"   Average Monthly:  {max_sales//3:>10,} units")

print(f"\n🎯 Needs Improvement: {channel_names[worst_channel]}")
print(f"   Quarterly Total:  {min_sales:>10,} units")
print(f"   Average Monthly:  {min_sales//3:>10,} units")

performance_gap = max_sales - min_sales
print(f"\n   Performance Gap:  {performance_gap:>10,} units")
print(f"   Gap Percentage:   {(performance_gap/max_sales)*100:.1f}%")

print("\n" + "=" * 70)
print("✅ TASK 3 COMPLETE")
print("=" * 70)
```

```python
import random

# [Use quarterly_data from Task 3]

print("\n" + "=" * 70)
print("TASK 4: PERFORMANCE ALERT SYSTEM")
print("=" * 70)

# ============================================
# DEFINE ALERT THRESHOLDS (Using Tuples)
# ============================================

# Threshold configuration tuple (name, value, description)
THRESHOLD_CONFIG = (
    ("sales_point_min", 500, "Minimum acceptable sales per point"),
    ("critical_low", 250, "Critical threshold requiring immediate action"),
    ("channel_min_percent", 20.0, "Minimum channel market share"),
    ("sku_min_units", 300, "Minimum units per SKU across all channels"),
    ("growth_target", 8.0, "Target month-over-month growth percentage")
)

# Convert to dictionary for easy access
THRESHOLDS = {}
for name, value, description in THRESHOLD_CONFIG:
    THRESHOLDS[name] = value

print("\n⚙️  Alert System Configuration:")
print("-" * 70)
for name, value, description in THRESHOLD_CONFIG:
    print(f"   {name:20} : {value:>8} ({description})")

# ============================================
# CALCULATE SALES POINT PERFORMANCE
# ============================================

print("\n" + "=" * 70)
print("🏬 SALES POINT PERFORMANCE ANALYSIS")
print("=" * 70)

# Dictionary to store performance by sales point
sales_point_performance = {}

# Analyze March data (most recent month)
march_sales = quarterly_data["March"]["sales"]

# Aggregate by sales point using while loop
sales_keys = list(march_sales.keys())
key_index = 0

while key_index < len(sales_keys):
    key = sales_keys[key_index]

    # Check if this is a GSM network sale (has sales point)
    if len(key) == 5 and key[0] == "GSM":
        channel, operator, distributor, point, sku = key

        # Initialize if not exists
        if point not in sales_point_performance:
            sales_point_performance[point] = {
                "total_units": 0,
                "operator": operator,
                "distributor": distributor
            }

        # Add units
        sales_point_performance[point]["total_units"] += march_sales[key]["units"]

    key_index += 1

# ============================================
# CLASSIFY SALES POINTS BY PERFORMANCE
# ============================================

# Initialize classification lists
critical_points = []
warning_points = []
good_points = []

# Classify using while loop
point_names = list(sales_point_performance.keys())
point_index = 0

while point_index < len(point_names):
    point_name = point_names[point_index]
    performance = sales_point_performance[point_name]
    units = performance["total_units"]

    # Classify based on thresholds
    if units < THRESHOLDS["critical_low"]:
        critical_points.append((point_name, units, performance))
    elif units < THRESHOLDS["sales_point_min"]:
        warning_points.append((point_name, units, performance))
    else:
        good_points.append((point_name, units, performance))

    point_index += 1

# ============================================
# DISPLAY CRITICAL ALERTS
# ============================================

print("\n🔴 CRITICAL ALERTS (Immediate Action Required):")
print("-" * 70)

if len(critical_points) > 0:
    # Sort by units (ascending) to show worst first
    critical_points.sort(key=lambda x: x[1])

    alert_index = 0
    while alert_index < len(critical_points):
        point, units, perf = critical_points[alert_index]
        deficit = THRESHOLDS["sales_point_min"] - units

        print(f"\n⚠️  {point}:")
        print(f"   Operator:     {perf['operator']}")
        print(f"   Distributor:  {perf['distributor']}")
        print(f"   Units Sold:   {units} (CRITICAL!)")
        print(f"   Below Target: {deficit} units")
        print(f"   Action:       🚨 Emergency review required")

        alert_index += 1
else:
    print("\n✅ No critical alerts - All points above critical threshold")

# ============================================
# DISPLAY WARNING ALERTS
# ============================================

print("\n🟡 WARNING ALERTS (Needs Attention):")
print("-" * 70)

if len(warning_points) > 0:
    warning_points.sort(key=lambda x: x[1])

    alert_index = 0
    while alert_index < len(warning_points):
        point, units, perf = warning_points[alert_index]
        deficit = THRESHOLDS["sales_point_min"] - units
        percent_of_target = (units / THRESHOLDS["sales_point_min"]) * 100

        print(f"\n📊 {point}:")
        print(f"   Operator:     {perf['operator']}")
        print(f"   Distributor:  {perf['distributor']}")
        print(f"   Units Sold:   {units}")
        print(f"   Target:       {THRESHOLDS['sales_point_min']}")
        print(f"   Achievement:  {percent_of_target:.1f}% of target")
        print(f"   Action:       📋 Performance improvement plan needed")

        alert_index += 1
else:
    print("\n✅ No warnings - All points meet minimum standards")

# ============================================
# DISPLAY GOOD PERFORMERS
# ============================================

print("\n🟢 TOP PERFORMERS:")
print("-" * 70)

if len(good_points) > 0:
    # Sort by units (descending) to show best first
    good_points.sort(key=lambda x: x[1], reverse=True)

    # Show top 5
    display_count = min(5, len(good_points))
    rank_index = 0

    while rank_index < display_count:
        point, units, perf = good_points[rank_index]
        above_target = units - THRESHOLDS["sales_point_min"]
        percent_above = (above_target / THRESHOLDS["sales_point_min"]) * 100

        print(f"\n⭐ #{rank_index + 1} {point}:")
        print(f"   Units Sold:   {units}")
        print(f"   Above Target: +{above_target} units ({percent_above:.1f}%)")

        rank_index += 1

print(f"\n   📊 Total Performing Well: {len(good_points)} locations")

# ============================================
# CHANNEL PERFORMANCE ALERTS
# ============================================

print("\n" + "=" * 70)
print("📊 CHANNEL PERFORMANCE ALERTS")
print("=" * 70)

# Get March data
march_totals = quarterly_data["March"]["totals"]
mar_grand_total = march_totals["Grand_Total"]

# Analyze each channel
channel_alerts = []

channels = ["Online", "Chain_Markets", "GSM_Network"]
channel_names = {
    "Online": "Online Stores",
    "Chain_Markets": "Chain Markets",
    "GSM_Network": "GSM Network"
}

channel_index = 0

while channel_index < len(channels):
    channel = channels[channel_index]
    channel_name = channel_names[channel]
    units = march_totals[channel]
    percentage = (units / mar_grand_total) * 100

    # Determine status
    if percentage < THRESHOLDS["channel_min_percent"]:
        status = "🔴 UNDERPERFORMING"
        action = "Increase marketing budget and sales incentives"
        priority = "HIGH"
    elif percentage < 30.0:
        status = "🟡 BELOW AVERAGE"
        action = "Review competitive positioning"
        priority = "MEDIUM"
    else:
        status = "🟢 STRONG"
        action = "Maintain current strategy"
        priority = "LOW"

    channel_alerts.append({
        "channel": channel_name,
        "status": status,
        "units": units,
        "percentage": percentage,
        "action": action,
        "priority": priority
    })

    channel_index += 1

# Display channel analysis
for alert in channel_alerts:
    print(f"\n{alert['channel']}:")
    print(f"   Status:      {alert['status']}")
    print(f"   Market Share: {alert['percentage']:.2f}%")
    print(f"   Units:       {alert['units']:,}")
    print(f"   Priority:    {alert['priority']}")
    print(f"   Action:      {alert['action']}")

# ============================================
# GROWTH RATE ALERTS
# ============================================

print("\n" + "=" * 70)
print("📈 GROWTH RATE ALERTS")
print("=" * 70)

feb_totals = quarterly_data["February"]["totals"]
feb_grand = feb_totals["Grand_Total"]

# Calculate month-over-month growth
mom_growth = ((mar_grand_total - feb_grand) / feb_grand) * 100

print(f"\n📊 February → March Growth: {mom_growth:+.2f}%")
print(f"   Target Growth:           {THRESHOLDS['growth_target']:.2f}%")

if mom_growth >= THRESHOLDS["growth_target"]:
    print(f"   Status:                  🟢 EXCEEDING TARGET")
    print(f"   Achievement:             {(mom_growth/THRESHOLDS['growth_target'])*100:.1f}% of target")
elif mom_growth >= THRESHOLDS["growth_target"] * 0.8:
    print(f"   Status:                  🟡 NEAR TARGET")
    print(f"   Achievement:             {(mom_growth/THRESHOLDS['growth_target'])*100:.1f}% of target")
else:
    print(f"   Status:                  🔴 BELOW TARGET")
    print(f"   Achievement:             {(mom_growth/THRESHOLDS['growth_target'])*100:.1f}% of target")
    print(f"   Action:                  Review sales strategies")

# ============================================
# ALERT SUMMARY
# ============================================

print("\n" + "=" * 70)
print("📋 ALERT SUMMARY REPORT")
print("=" * 70)

total_alerts = len(critical_points) + len(warning_points)

print(f"\n📊 Total Active Alerts: {total_alerts}")
print(f"\n   🔴 Critical Sales Points:  {len(critical_points)}")
print(f"   🟡 Warning Sales Points:   {len(warning_points)}")
print(f"   🟢 Performing Well:        {len(good_points)}")

# Overall assessment using while loop
if total_alerts == 0:
    overall_status = "EXCELLENT"
    overall_emoji = "🎉"
    overall_action = "No immediate action required"
elif total_alerts < 5:
    overall_status = "GOOD"
    overall_emoji = "✅"
    overall_action = "Address flagged items in routine review"
elif total_alerts < 10:
    overall_status = "MODERATE CONCERN"
    overall_emoji = "🟡"
    overall_action = "Schedule focused review of underperforming areas"
else:
    overall_status = "URGENT ATTENTION NEEDED"
    overall_emoji = "🔴"
    overall_action = "Immediate management intervention required"

print(f"\n{overall_emoji} Overall Status: {overall_status}")
print(f"   Recommended Action: {overall_action}")

# ============================================
# ACTION ITEMS GENERATOR
# ============================================

print("\n" + "=" * 70)
print("✅ RECOMMENDED ACTION ITEMS")
print("=" * 70)

action_items = []

# Generate actions for critical points
if len(critical_points) > 0:
    item_index = 0
    while item_index < min(3, len(critical_points)):
        point, units, perf = critical_points[item_index]

        action_items.append({
            "priority": "🔴 HIGH",
            "area": f"Sales Point: {point}",
            "issue": f"Only {units} units sold (Critical threshold)",
            "action": "Emergency performance review",
            "responsible": "Regional Sales Manager",
            "deadline": "Within 3 days"
        })

        item_index += 1

# Generate actions for warnings
if len(warning_points) > 0:
    action_items.append({
        "priority": "🟡 MEDIUM",
        "area": f"Multiple Sales Points ({len(warning_points)} locations)",
        "issue": "Below target performance",
        "action": "Implement performance improvement plan",
        "responsible": "District Managers",
        "deadline": "Within 2 weeks"
    })

# Generate actions for channel issues
for alert in channel_alerts:
    if alert["priority"] == "HIGH":
        action_items.append({
            "priority": "🟡 MEDIUM",
            "area": f"Channel: {alert['channel']}",
            "issue": f"Market share only {alert['percentage']:.1f}%",
            "action": alert["action"],
            "responsible": "Marketing Team",
            "deadline": "Within 1 month"
        })

# Display action items
if len(action_items) == 0:
    print("\n🎉 No action items required - All metrics healthy!")
else:
    action_index = 0
    while action_index < len(action_items):
        item = action_items[action_index]

        print(f"\n{action_index + 1}. [{item['priority']}] {item['area']}")
        print(f"   Issue:        {item['issue']}")
        print(f"   Action:       {item['action']}")
        print(f"   Responsible:  {item['responsible']}")
        print(f"   Deadline:     {item['deadline']}")

        action_index += 1

# ============================================
# EMAIL ALERT SIMULATION
# ============================================

print("\n" + "=" * 70)
print("📧 AUTOMATED ALERT NOTIFICATION")
print("=" * 70)

print("\n" + "-" * 70)
print("TO:       management@techsphere.com")
print("FROM:     sales-monitoring@techsphere.com")
print("SUBJECT:  [ALERT] Performance Monitoring Report - March")
print("DATE:     " + "2024-03-31")
print("PRIORITY: " + ("HIGH" if total_alerts >= 5 else "NORMAL"))
print("-" * 70)

print("\nDear Management Team,\n")
print("The automated sales monitoring system has completed the")
print("monthly performance analysis and identified the following:\n")

if len(critical_points) > 0:
    print(f"🔴 URGENT: {len(critical_points)} sales point(s) performing critically low")
if len(warning_points) > 0:
    print(f"🟡 ATTENTION: {len(warning_points)} sales point(s) below target")

if total_alerts == 0:
    print("✅ All performance metrics are meeting or exceeding targets")

print(f"\nTotal Action Items: {len(action_items)}")
print("\nPlease review the detailed report and take appropriate action.")
print("The complete analysis is available in the sales dashboard.")

print("\nBest regards,")
print("TechSphere Automated Monitoring System")
print("-" * 70)

print("\n" + "=" * 70)
print("✅ TASK 4 COMPLETE")
print("=" * 70)
```
